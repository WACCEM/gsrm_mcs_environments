### Read stats 

```python
main_dir = '/pscratch/sd/p/paccini/temp/hackathon/updated_environmental_variables/'
mcs_stats = 'df_filtered_mcs_track_ids/'
# main_dir = '/Users/pacc275/local_documents/hackathon/updated_environmental_variables/' #local
# --- Define main components ---
TRACK_MODELS = ['um', 'icon', 'scream', 'ifs', 'nicam', 'obs', 'obs_v7'] 
VARIABLES = ['prw', 'rh500', 'rh850', 'q500', 'q850', 'omega500', 'omega850',  'lShear', 'dShear'] #'w850', 'w500',
CRITERIA = ['all', 'min2']
```

```python
# --- Path Configuration ---

# 1. Map your track model name (e.g., 'obs') to the 
#    corresponding environmental data directory (e.g., 'era5')
ENV_MODEL_DIR_MAP = {
    'um': 'UM_all',
    'icon': 'ICON_all',
    'scream': 'SCREAM_all',
    'ifs': 'IFS_all',
    'nicam': 'NICAM_all',
    'obs_v7': 'ERA5_IMERGv7_all' ,
    'obs': 'era5'  # Key insight: 'obs' tracks map to 'era5' env data
}

# 2. Define the path pattern for each variable.
#    'type': 'dir' (read the whole dir as parquet) or 
#            'glob' (find a single file using a wildcard '*')
VAR_PATH_PATTERNS = {
    # Default patterns used by most models (um, scream, etc.)
    'default': {
        'prw':      {'type': 'dir', 'pattern': 'prw/'},
        'rh500':    {'type': 'glob', 'pattern': 'hur/mcs_env_hur_*_hur_500hPa.parquet'},
        'rh850':    {'type': 'glob', 'pattern': 'hur/mcs_env_hur_*_hur_850hPa.parquet'},
        'q500':     {'type': 'glob', 'pattern': 'hus/mcs_env_hus_*_hus_500hPa.parquet'},
        'q850':     {'type': 'glob', 'pattern': 'hus/mcs_env_hus_*_hus_850hPa.parquet'},
        'omega500': {'type': 'glob', 'pattern': 'omega/mcs_env_omega_*_omega_500hPa.parquet'},
        'omega850': {'type': 'glob', 'pattern': 'omega/mcs_env_omega_*_omega_850hPa.parquet'},
        # 'w500':     {'type': 'glob', 'pattern': 'wa/mcs_env_wa_*_wa_500hPa.parquet'},
        # 'w850':     {'type': 'glob', 'pattern': 'wa/mcs_env_wa_*_wa_850hPa.parquet'},
        'lShear':   {'type': 'dir', 'pattern': 'low_shear_magnitude/'},
        'dShear':   {'type': 'dir', 'pattern': 'deep_shear_magnitude/'},
    },
    # Overrides for the 'obs' (era5) model
    'obs': {
        'q500':     {'type': 'dir', 'pattern': 'q_500mb/'},
         'rh500':  {'type': 'dir', 'pattern': 'rh_500mb/'},
        'q850':     {'type': 'dir', 'pattern': 'q_850mb/'},
         'rh850':  {'type': 'dir', 'pattern': 'rh_850mb/'},
          'prw': {'type': 'dir', 'pattern':'TCWV/'},
        'rh500':  {'type': 'dir', 'pattern': 'rh_500mb/'},
        'omega500':     {'type': 'dir', 'pattern': 'w_500mb/'},
        'omega850':     {'type': 'dir', 'pattern': 'w_850mb/'},
        # 'lShear':   {'type': 'dir', 'pattern': 'shear_mag_0to2km/'},
        # 'dShear':   {'type': 'dir', 'pattern': 'shear_mag_0to10km/'},
    }
}

# --- New Path-Finding Function ---

def get_env_path(base_out_dir, model_name, var_name):
    """
    Finds the correct file path for a given model and variable,
    handling inconsistent names and date stamps.
    
    'model_name' should be from TRACK_MODELS (e.g., 'um', 'obs')
    """
    
    # 1. Get the top-level directory name (e.g., 'UM_all' or 'era5')
    if model_name not in ENV_MODEL_DIR_MAP:
        raise KeyError(f"Model '{model_name}' not defined in ENV_MODEL_DIR_MAP.")
    model_dir = ENV_MODEL_DIR_MAP[model_name]
    
    # 2. Get the path pattern rule for this variable
    # Check for a model-specific override first (e.g., for 'obs')
    if model_name in VAR_PATH_PATTERNS and var_name in VAR_PATH_PATTERNS[model_name]:
        rule = VAR_PATH_PATTERNS[model_name][var_name]
    # Otherwise, use the default rule
    elif var_name in VAR_PATH_PATTERNS['default']:
        rule = VAR_PATH_PATTERNS['default'][var_name]
    else:
        raise KeyError(f"Variable '{var_name}' not defined in VAR_PATH_PATTERNS.")
        
    pattern_type = rule['type']
    pattern = rule['pattern']
    
    # 3. Construct the full search path
    full_search_path = os.path.join(base_out_dir, model_dir, pattern)
    
    # 4. Find the file/directory
    if pattern_type == 'dir':
        # For 'dir' type, we just return the path as-is.
        return full_search_path
        
    elif pattern_type == 'glob':
        # For 'glob', we find all files matching the pattern
        search_results = glob.glob(full_search_path)
        
        if len(search_results) == 0:
            raise FileNotFoundError(f"No file found for {model_name}/{var_name}. "
                                    f"Searched: {full_search_path}")
        if len(search_results) > 1:
            # This is ambiguous, which is bad.
            raise FileNotFoundError(f"Multiple files found for {model_name}/{var_name}. "
                                     f"Searched: {full_search_path}. "
                                     f"Found: {search_results}")
        
        # Success! Return the one and only match.
        return search_results[0]
```

```python
# --- Transformation Configuration ---
# Define all model/variable-specific unit conversions here

TRANSFORM_CONFIG = {
    # 'scream': {
    #     'rh500': lambda df: df.assign(mean=df['mean'] * 100),
    #     'rh850': lambda df: df.assign(mean=df['mean'] * 100),
    # },
    'nicam': {
        'rh500': lambda df: df.assign(mean=df['mean'] * 100),
        'rh850': lambda df: df.assign(mean=df['mean'] * 100),
    },
    'obs': {  # 'obs' is our alias for 'era5imergv6'
        'q500': lambda df: df.assign(mean=df['mean'] / 1000),
        'q850': lambda df: df.assign(mean=df['mean'] / 1000),
    },

}
```

```python
# --- Load Environmental Data ---
raw_env_data = {}
for model in TRACK_MODELS: # Loop over our base track models
    raw_env_data[model] = {}
    for var in VARIABLES:
        try:
            # 1. Find the path (same as before)
            path = get_env_path(main_dir, model, var)
            
            # 2. Load the data
            df = pd.read_parquet(path)
            
            # 3. *** NEW: Apply Transformation ***
            # Get the specific transform function from our config
            # If no rule exists, it defaults to a function that does nothing (lambda x: x)
            transform_func = TRANSFORM_CONFIG.get(model, {}).get(var, lambda x: x)
            
            # Apply the function to the DataFrame
            df_transformed = transform_func(df)
            
            # 4. Store the standardized DataFrame
            raw_env_data[model][var] = df_transformed
        #     # Our function handles all the ugly logic!
        #     path = get_env_path(main_dir, model, var)
        #     raw_env_data[model][var] = pd.read_parquet(path)
        except Exception as e:
            # Catch errors from loading OR transforming
            print(f"Warning: Could not load env data for {model} - {var}. Error: {e}")
            raw_env_data[model][var] = None

# --- Load Track Data ---
raw_track_data = {}
for model in TRACK_MODELS:
    raw_track_data[model] = {}
    for crit in CRITERIA:
        try:
            crit_suffix = f'_{crit}' if crit != 'all' else ''
            # path = os.path.join(mcs_stats, f'df_trop_{model}{crit_suffix}.pkl')
            path = os.path.join(mcs_stats, f'with_circular_lffiltered_{model}{crit_suffix}_r2.pkl')
            raw_track_data[model][crit] = pd.read_pickle(path)
        except Exception as e:
            print(f"Warning: Could not load track data for {model} - {crit}. Error: {e}")
            raw_track_data[model][crit] = None
```

```python
# --- Filtering environments ---
# def get_filtered_df(df_model, df_tracks, model_track='track_id', filter_track='track_id'):
#     if df_model is None or df_tracks is None:
#         return None
#     filtered_df = df_model[df_model[model_track].isin(df_tracks[filter_track])]
#     return filtered_df
def get_filtered_df(df_main, df_tracks, lf_thresh=0.1,lf_var='total_land_fraction_track'):
    """
    Merges main data with track data data and filters by circular land fraction threshold.
    """
    # Safe Merge (Inner Join) on track_id to ensure data integrity
    # Adjust 'track_id' if your column names differ between dataframes
    merged_df = pd.merge( df_main, df_tracks[['track_id', 'total_total_rain',lf_var]], on='track_id', how='inner')
    
    # Filter by Land Fraction Threshold
    # Ensure it handles NaNs if necessary, though inner join usually drops non-matches
    final_df = merged_df[merged_df[lf_var] <= lf_thresh].copy()
    
    return final_df

def remove_longdurations(df_model, feature='duration_hours', p_high=0.9):
    if df_model is None:
        return None
    q_high = df_model[feature].quantile(p_high)
    df_filtered = df_model[(df_model[feature] <= q_high)].copy()
    return df_filtered
```

```python
# --- Final processing loop  ---
processed_data = {}

for model in TRACK_MODELS:
    processed_data[model] = {}
    for crit in CRITERIA:
        base_track_df = raw_track_data[model].get(crit)
        
        if base_track_df is None:
            print(f"Skipping {model} - {crit}: Missing base track data.")
            continue
            
        long_removed_track_df = remove_longdurations(base_track_df)
        
        crit_key_base = crit
        crit_key_longremoved = f'{crit}_longremoved'
        
        processed_data[model][crit_key_base] = {}
        processed_data[model][crit_key_longremoved] = {}
        
        for var in VARIABLES:
            raw_env_df = raw_env_data[model].get(var)
            
            if raw_env_df is None:
                # This will now correctly skip missing 'era5' data
                continue 

            processed_data[model][crit_key_base][var] = get_filtered_df(
                raw_env_df, base_track_df
            )
            
            processed_data[model][crit_key_longremoved][var] = get_filtered_df(
                raw_env_df, long_removed_track_df
            )

print("Processing complete!")
```

```python
# Select variables to plot
vars_to_plot = ['rh850', 'q850', 'prw', 'omega850']

# Create subplots (2x2 grid)
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(8, 6), constrained_layout=True)
axes = axes.flatten() # Flattens the 2x2 grid into a 1D list for easy looping

# --- Outer Loop: Iterate through Variables (Subplots) ---
for ax, var in zip(axes, vars_to_plot):
    
    # Get settings for this variable
    if var not in VAR_SETTINGS:
        print(f"Skipping {var}: No bin settings defined.")
        continue
        
    bins = VAR_SETTINGS[var]['bins']
    mid_bins = 0.5 * (bins[1:] + bins[:-1]) # Calculate x-axis midpoints
    
    # --- Inner Loop: Iterate through Models (Lines) ---
    for model in MODEL_STYLES.keys():
        
        # 1. Retrieve Data
        # Safely get the dataframe from your nested dictionary
        try:
            df = processed_data[model][CRIT][var]
        except KeyError:
            continue # Skip if data doesn't exist for this model
            
        if df is None: continue

        # 2. Filter Data (Radius & Time)
        subset = df[(df.radius == RR) & (df.time_offset_hours == TT)]
        
        if subset.empty:
            continue

        # 3. Calculate Histogram
        # Using numpy vectorization is faster than list comprehension
        counts, _ = np.histogram(subset['mean'].values, bins=bins, density=False)
        
        # Avoid division by zero if sum is 0
        total_counts = counts.sum()
        if total_counts > 0:
            percentages = (counts / total_counts) * 100
        else:
            percentages = np.zeros_like(counts)

        # 4. Plot
        style = MODEL_STYLES[model]
        ax.plot(mid_bins, percentages, 
                color=style['color'], 
                label=style['label'], 
                linewidth=1.5,
                linestyle=style.get('ls', '-')) # Default to solid line

    # --- Subplot Formatting ---
    ax.set_xlabel(VAR_SETTINGS[var]['xlabel'], fontsize=12)
    ax.set_ylabel('Frequency (%)', fontsize=12)
    ax.grid(True, alpha=0.3)
    ax.set_facecolor('white')
    ax.grid(color='k', linestyle='--', linewidth=0.3, alpha=0.3)
    ax.spines['left'].set_color('k'); ax.spines['bottom'].set_color('k')
    ax.spines['left'].set_linewidth(.8); ax.spines['bottom'].set_linewidth(.8)
    
    # Only add legend to the first plot to avoid clutter (or add to all if preferred)
    if var == vars_to_plot[0]:
        ax.legend(fontsize=9, loc='upper left',)

plt.suptitle(f"Distributions at Radius={RR}, Time={TT}h", fontsize=11)
plt.show()
```

### Choosing similar environments for conditional probability distributions

```python
## 1. Objectively calculate overlapping environmental conditions
def get_common_overlap(data_dict, models, crit='all', var='prw', low_p=10, high_p=90):
    """
    Calculates the intersection of ranges [P_low, P_high] across all models.
    """
    global_min = -np.inf
    global_max = np.inf
    
    ranges = {}
    
    for model in models:
        # --- FIX: Access dictionary dynamically using 'crit' ---
        try:
            df = data_dict[model][crit][var]
        except KeyError:
            print(f"Warning: Data not found for {model} -> {crit} -> {var}")
            return None, None

        if df is None:
            print(f"Warning: DataFrame is None for {model} -> {crit} -> {var}")
            return None, None
        
        # Extract values for the CONTROL variable (e.g., PRW)
        values = df['mean'].values
        
        # Calculate percentiles
        lower = np.percentile(values, low_p)
        upper = np.percentile(values, high_p)
        ranges[model] = (lower, upper)
        
        # Update global intersection
        global_min = max(global_min, lower)
        global_max = min(global_max, upper)
    
    # Check if a valid overlap exists
    if global_min >= global_max:
        print(f"No overlap found! Min of tops ({global_max:.2f}) < Max of bottoms ({global_min:.2f})")
        return None, ranges
        
    return (global_min, global_max), ranges
```

```python
### 2. Total precipitation amount distribution in the "common-range" environmental conditions
def plot_rain_in_common_env(data_dict, models, crit='all', 
                            control_var='prw', overlap=None,
                            target_var='total_total_rain', 
                            radius=2, time=0):
    if overlap==None:
    # 1. Get Objective Range using the corrected function
        overlap, ranges = get_common_overlap(data_dict, models, crit=crit, var=control_var)
    
    # if not overlap:
    #     return
    
    print(f"Comparison Criteria: {crit}")
    print(f"Common {control_var} range: {overlap[0]:.2f} - {overlap[1]:.2f}")

    # 2. Collect Data
    plot_data = []
    
    for model in models:
        # --- FIX: Access dictionary dynamically ---
        try:
            df = data_dict[model][crit][control_var]
        except KeyError:
            continue

        if df is None: continue
        
        # Apply Filters:
        # 1. Radius & Time
        # 2. Control Variable Range (e.g. PRW is within the overlap)
        mask = (
            (df['radius'] == radius) & 
            (df['time_offset_hours'] == time) & 
            (df['mean'] >= overlap[0]) & 
            (df['mean'] <= overlap[1])
        )
        
        subset = df[mask].copy()
        
        if subset.empty:
            print(f"Warning: {model} has no data in the common range.")
            continue
        
        # Keep only the target rain data and label it
        subset = subset[[target_var]].rename(columns={target_var: 'rain'})
        subset['Model'] = MODEL_STYLES[model]['label'] 
        plot_data.append(subset)
    
    if not plot_data:
        print("No data available to plot.")
        return

    full_df = pd.concat(plot_data)
    
    # 3. Plot
    plt.figure(figsize=(7, 4))
    
    # KDE Plot
    sns.kdeplot(data=full_df, x='rain', hue='Model', 
                palette={v['label']: v['color'] for k,v in MODEL_STYLES.items() if k in models},
                linewidth=2.5, common_norm=False)
    # sns.move_legend(title=None, fontsize=8)
    plt.title(f"Rain Dist. ({crit})\nwhere {overlap[0]:.1f} < {control_var} < {overlap[1]:.1f}")
    plt.xlabel("Total Rainfall (mm)"); plt.xscale('log')
    plt.xlim(left=0) # Lock left axis to 0 for rain
    plt.grid(True, alpha=0.3)
    plt.show()

# --- RUN CONFIGURATION ---
# Case: Similar Distributions (Excluding SCREAM/UM outliers for PRW)
# p1. Define the models you want to compare
# selected_models = ['icon', 'ifs', 'obs', 'nicam'] 

# # 2. Run the plot with your specific criteria
# plot_rain_in_common_env(
#     processed_data, 
#     selected_models, 
#     crit='min2_longremoved',  # <--- Now handled correctly!
#     control_var='prw',
#     target_var='total_total_rain'
# )
```

```python
def plot_low_high_regime(data_dict, models, crit='all', 
                            control_var='prw', target_var='total_total_rain',
                            radius=2, time=0):
    
    # 1. Get Common Overlap Range
    overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=control_var)
    if not overlap: return
    
    print(f"Common {control_var} range: {overlap[0]:.2f} - {overlap[1]:.2f}")

    # 2. Collect Data & Calculate Shared Median
    # We need to collect all valid data first to find the median of the OVERLAP region
    all_valid_values = []
    model_data_store = {}

    for model in models:
        try:
            df = data_dict[model][crit][control_var]
        except KeyError: continue
        if df is None: continue

        # Filter: Radius, Time, and Overlap Range
        mask = (
            (df.radius == radius) & 
            (df.time_offset_hours == time) & 
            (df['mean'] >= overlap[0]) & 
            (df['mean'] <= overlap[1])
        )
        valid_data = df[mask].copy()
        model_data_store[model] = valid_data
        all_valid_values.extend(valid_data['mean'].values)

    if not all_valid_values:
        print("No data found in overlap region.")
        return

    # Calculate split point (median of the constrained environment)
    split_val = np.median(all_valid_values)
    print(f"Splitting Regimes at {control_var} = {split_val:.2f}")

    # 3. Plotting
    fig, axes = plt.subplots(1, 2, figsize=(9, 4), sharey=False, sharex=True)
    titles = [f"Low {control_var} (<{split_val:.1f})", f"High {control_var} (>{split_val:.1f})"]

    for i, ax in enumerate(axes):
        plot_dfs = []
        
        for model in models:
            if model not in model_data_store: continue
            
            df = model_data_store[model]
            
            # Filter Low vs High
            if i == 0: 
                subset = df[df['mean'] <= split_val]
            else:      
                subset = df[df['mean'] > split_val]
            
            # Create Label with Count (N)
            count = len(subset); 
            label_str = f"{MODEL_STYLES[model]['label']} \n(N={count})"
            
            # Prepare for seaborn
            temp = subset[[target_var]].copy()
            temp.columns = ['rain']
            temp['Model_Label'] = label_str
            temp['Model_Key'] = model # Keep key for color mapping
            plot_dfs.append(temp)
        
        if not plot_dfs: continue
        full_df = pd.concat(plot_dfs)
        
        # Create custom palette that maps the NEW labels to the OLD colors
        # We map the 'Model_Label' back to the color defined in MODEL_STYLES[model]
        unique_labels = full_df[['Model_Label', 'Model_Key']].drop_duplicates()

        # 3. Plot
        palette = {row.Model_Label: MODEL_STYLES[row.Model_Key]['color'] for _, row in unique_labels.iterrows()}
        # KDE Plot
        sns.kdeplot(data=full_df, x='rain', hue='Model_Label', 
                    palette=palette,ax=ax,
                    linewidth=2.5, common_norm=False)

        sns.move_legend(ax, "upper right", title=None, fontsize=8)
        # ax.legend(fontsize=10)
        ax.set_title(titles[i]); ax.set_xscale('log')
        ax.set_xlabel("Total Rain (mm)")
        
    plt.tight_layout()
    plt.show()
```

```python
def plot_rain_phase_space_normalized(data_dict, models, crit='all',
                                     constrain_var='prw', 
                                     x_var='omega850', 
                                     target_var='total_total_rain',n_y_bins=3,
                                     norm_method='quantile', # 'quantile' or 'minmax'
                                     p_low=0.05, p_high=0.95,cmap=mycmap.whiteTOblueTOgreen):
    
    # 1. Get Constraints and Global Ranges
    overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=constrain_var)
    if not overlap: return

    # To ensure bins are identical across subplots, we need Global Min/Max for the X-variable
    # We scan all models briefly to find the min/max of Omega within the PRW overlap
    x_vals_all = []
    y_vals_all = []
    for m in models:
        try:
            df_c = data_dict[m][crit][constrain_var]
            df_x = data_dict[m][crit][x_var]

            # Filter for common track_ids in the overlap range
            # (This assumes track_ids align, simplified for speed)
            valid_c = df_c[df_c['mean'].between(overlap[0], overlap[1])]
            valid_tracks = valid_c['track_id']
            
            # Store Y values (Constraint) strictly within overlap
            y_vals_all.extend(valid_c['mean'].values)
            
            # Rough filter to get relevant range
    
            relevant_x = df_x[df_x['track_id'].isin(valid_tracks)]
            x_vals_all.extend(relevant_x['mean'].values)
        except: continue
            
    if not x_vals_all or not y_vals_all: 
        print("No data found for binning.")
        return
    
    # Define Global Bins
    # Y-Bins: Percentiles of the data within the overlap
    # If n=2, quantiles=[0, 0.5, 1.0]. If n=3, quantiles=[0, 0.33, 0.66, 1.0]
    qs = np.linspace(0, 1, n_y_bins + 1)
    y_bins = np.unique(np.quantile(y_vals_all, qs)) # Unique handles cases with low precision

    # X-Bins: The Omega range divided into 5
    x_bins = np.linspace(np.percentile(x_vals_all, 1), np.percentile(x_vals_all, 99), 6) # 5 bins (cutting top/bottom 1% outliers)

    # 2. Prepare Subplots
    n_models = len(models)
    cols = 3
    rows = (n_models // cols) + (1 if n_models % cols > 0 else 0)
    fig, axes = plt.subplots(rows, cols, figsize=(5*cols, 4*rows), constrained_layout=True)
    axes = axes.flatten()

    mappable = None
    # 3. Loop Models
    for idx, model in enumerate(models):
        ax = axes[idx]
        
        try:
            # Load Dataframes
            df_c = data_dict[model][crit][constrain_var] # PRW
            df_x = data_dict[model][crit][x_var]         # Omega
            
            # Filter Radius/Time (Assumed 2/0 based on previous context, or pass as args)
            f_c = df_c[(df_c.radius==2) & (df_c.time_offset_hours==0)]
            f_x = df_x[(df_x.radius==2) & (df_x.time_offset_hours==0)]
            
            # Merge: C + X + T
            merged = pd.merge(f_c[['track_id','mean',target_var]], f_x[['track_id','mean',target_var]], on=['track_id',target_var], suffixes=('_c', '_x'))
            
            # Filter by Constraint Overlap
            data = merged[(merged['mean_c'] >= overlap[0]) & (merged['mean_c'] <= overlap[1])].copy()
            
            if data.empty:
                ax.text(0.5, 0.5, "No Overlap Data", ha='center', va='center')
                continue

            # Binning
            data['x_bin'] = pd.cut(data['mean_x'], bins=x_bins, include_lowest=True)
            data['y_bin'] = pd.cut(data['mean_c'], bins=y_bins, include_lowest=True)
            
            # Aggregate: Calculate Mean Rain per cell
            # We use unstack to create a matrix for the heatmap
            pivot_mean = data.groupby(['y_bin', 'x_bin'], observed=False)[target_var].mean().unstack()
            pivot_count = data.groupby(['y_bin', 'x_bin'], observed=False)['track_id'].count().unstack().fillna(0).astype(int)
            
            # --- NORMALIZATION (User Logic) ---
            # Flatten to apply normalization logic
            data_flat = pivot_mean.values
            
            if norm_method == 'quantile':
                vmin_model = np.nanquantile(data_flat, p_low)
                vmax_model = np.nanquantile(data_flat, p_high)
            else: # minmax
                vmin_model = np.nanmin(data_flat)
                vmax_model = np.nanmax(data_flat)
            
            # Normalize to 0-1 range
            denom = vmax_model - vmin_model
            if denom > 0:
                norm_data = (data_flat - vmin_model) / denom
                norm_data = np.clip(norm_data, 0, 1)
            else:
                norm_data = np.zeros_like(data_flat) # or 0.5 if flat
            
            # Custom colormap
            levels = MaxNLocator(nbins=20).tick_values(0, 1)
            norm = BoundaryNorm(levels, ncolors=plt.get_cmap(cmap).N, clip=True)
            
            # Reconstruct DataFrame for plotting
            norm_df = pd.DataFrame(norm_data, index=pivot_mean.index, columns=pivot_mean.columns)
            
            # Sort index descending so high PRW is at top of Y-axis
            norm_df = norm_df.sort_index(ascending=False) # High PRW on top

            # Apply the same sorting and shape to the Count matrix
            pivot_count_sorted = pivot_count.loc[norm_df.index, norm_df.columns]

            # Plot Heatmap
            im = sns.heatmap(norm_df, ax=ax, cmap=cmap, norm=norm,
                        cbar=False, cbar_kws={'label': 'Norm. Rain Intensity'},
                        vmin=0, vmax=1, annot=pivot_count_sorted, fmt="d", annot_kws={"size": 8})
            # Capture the mappable object from the Axes object
            if idx == 0:
                mappable = im.collections[0]
            
            ax.set_title(f"{MODEL_STYLES[model]['label']}\n(N={len(data)})")
            ax.set_xlabel(x_var)
            ax.set_ylabel(constrain_var)
            
            # Clean up tick labels for readability
            ax.set_xticklabels([f"{b.mid:.2f}" for b in norm_df.columns], rotation=45)
            ax.set_yticklabels([f"{b.mid:.1f}" for b in norm_df.index], rotation=0)

        except Exception as e:
            print(merged.columns)
            print(f"Error plotting {model}: {e}")
            ax.set_visible(False)

    # Turn off unused subplots
    for i in range(len(models), len(axes)):
        axes[i].axis('off')
    # Global Colorbar Application    
    if mappable:
        cbar_label = f"{'Normalized '}{target_var}"
        fig.colorbar(mappable, ax=axes.tolist(), orientation='vertical',label=cbar_label, 
                     shrink=0.8, pad=0.02, aspect=30)
            
        plt.suptitle(f"Norm. Rain vs {x_var} | Constrained {constrain_var}: {overlap[0]:.1f}-{overlap[1]:.1f}mm", y=1.02)
        plt.show()

# --- HOW TO RUN ---
# selected = ['scream', 'um', 'icon', 'ifs', 'nicam', 'obs']
# plot_rain_phase_space_normalized(processed_data, selected, crit='min2_longremoved')
```

```python
def plot_rain_phase_space_normalized(data_dict, models, crit='all',
                                     constrain_var='prw', 
                                     x_var='omega850', 
                                     target_var='total_total_rain', n_y_bins=3,
                                     n_x_bins=5, # New parameter for consistent X-bin count
                                     norm_method='quantile', 
                                     p_low=0.05, p_high=0.95, cmap=mycmap.whiteTOblueTOgreen): 
    
    # 1. Get Constraints and Global Ranges (PRW overlap)
    overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=constrain_var)
    if not overlap: return

    # Collect Y values only for GLOBAL Y-bin calculation
    y_vals_all = []
    for m in models:
        try:
            df_c = data_dict[m][crit][constrain_var]
            valid_c = df_c[df_c['mean'].between(overlap[0], overlap[1])]
            y_vals_all.extend(valid_c['mean'].values)
        except: continue
            
    if not y_vals_all: return
    
    # Define Global Y-Bins (PRW - based on percentiles, consistent across models)
    qs = np.linspace(0, 1, n_y_bins + 1)
    y_bins = np.unique(np.quantile(y_vals_all, qs))

    # 2. Prepare Subplots and Global Colorbar Setup
    n_models = len(models)
    cols = 3
    rows = (n_models // cols) + (1 if n_models % cols > 0 else 0)
    fig, axes = plt.subplots(rows, cols, figsize=(4*cols, 3*rows), sharey=True,constrained_layout=True)
    axes = axes.flatten()

    # Define global colorbar norm based on the final normalized range [0, 1]
    levels = MaxNLocator(nbins=20).tick_values(0, 1)
    # Ensure 'cmap' is a valid Matplotlib Colormap object or string
    cmap_obj = plt.get_cmap(cmap) if isinstance(cmap, str) else cmap 
    global_norm = BoundaryNorm(levels, ncolors=cmap_obj.N, clip=True)

    mappable = None
    
    # 3. Loop Models (Main Logic)
    for idx, model in enumerate(models):
        ax = axes[idx]
        
        try:
            # Load Dataframes
            df_c = data_dict[model][crit][constrain_var] 
            df_x = data_dict[model][crit][x_var]          
            
            # Filter Radius/Time & Merge
            f_c = df_c[(df_c.radius==2) & (df_c.time_offset_hours==0)]
            f_x = df_x[(df_x.radius==2) & (df_x.time_offset_hours==0)]
            
            # Merge: C + X + T
            merged = pd.merge(f_c[['track_id','mean',target_var]], f_x[['track_id','mean',target_var]], 
                             on=['track_id',target_var], suffixes=('_c', '_x'))
            
            # Filter by Constraint Overlap
            data = merged[(merged['mean_c'] >= overlap[0]) & (merged['mean_c'] <= overlap[1])].copy()
            
            if data.empty:
                ax.text(0.5, 0.5, "No Overlap Data", ha='center', va='center')
                continue

            # --- LOCAL OMEGA BINNING (X-axis) ---
            # Calculate Omega bins based ONLY on the current model's data
            omega_vals = data['mean_x'].values
            x_edges = np.linspace(np.percentile(omega_vals, 1), np.percentile(omega_vals, 99), n_x_bins + 1)
            
            # Binning
            data['x_bin'] = pd.cut(data['mean_x'], bins=x_edges, include_lowest=True)
            data['y_bin'] = pd.cut(data['mean_c'], bins=y_bins, include_lowest=True)
            
            # Aggregate Mean and Count
            pivot_mean = data.groupby(['y_bin', 'x_bin'], observed=False)[target_var].mean().unstack()
            pivot_count = data.groupby(['y_bin', 'x_bin'], observed=False)['track_id'].count().unstack().fillna(0).astype(int)
            
            # --- NORMALIZATION (Per-Model) ---
            data_flat = pivot_mean.values
            if norm_method == 'quantile':
                vmin_model = np.nanquantile(data_flat, p_low)
                vmax_model = np.nanquantile(data_flat, p_high)
            else: # minmax
                vmin_model = np.nanmin(data_flat)
                vmax_model = np.nanmax(data_flat)
            
            denom = vmax_model - vmin_model
            if denom > 0:
                norm_data = (data_flat - vmin_model) / denom
                norm_data = np.clip(norm_data, 0, 1)
            else:
                norm_data = np.zeros_like(data_flat)
                
            # Reconstruct and Sort
            norm_df = pd.DataFrame(norm_data, index=pivot_mean.index, columns=pivot_mean.columns)
            norm_df = norm_df.sort_index(ascending=False)
            pivot_count_sorted = pivot_count.loc[norm_df.index, norm_df.columns]

            # Plot Heatmap
            im = sns.heatmap(norm_df, ax=ax, cmap=cmap_obj, norm=global_norm,
                             cbar=False, annot=pivot_count_sorted, fmt="d", annot_kws={"size": 8})
            
            # Capture the mappable object
            if idx == 0:
                mappable = im.collections[0]
            
            ax.set_title(f"{MODEL_STYLES[model]['label']}\n(Total N={len(data)})")
            ax.set_xlabel(x_var)
            ax.set_ylabel(constrain_var)
            
            # Clean up tick labels
            ax.set_xticklabels([f"{b.mid:.2f}" for b in norm_df.columns], rotation=45)
            y_labels = [f"{b.mid:.1f}" for b in norm_df.index]
            ax.set_yticklabels(y_labels, rotation=0)

        except Exception as e:
            # Note: The merge on 'track_id', 'target_var' might be causing issues if target_var is not identical
            # in both merged dataframes. I simplified it to merge on just 'track_id'.
            print(f"Error plotting {model}: {e}")
            ax.set_visible(False)

    # Turn off unused subplots
    for i in range(len(models), len(axes)):
        axes[i].axis('off')
        
    # Global Colorbar Application
    if mappable:
        cbar_label = f"Normalized {target_var} [0 to 1]"
        fig.colorbar(mappable, ax=axes.tolist(), orientation='vertical', label=cbar_label,
                     shrink=0.8, pad=0.02, aspect=30)
        
    plt.suptitle(f"Normalized Rain vs {x_var} | Constrained {constrain_var}: {overlap[0]:.1f}-{overlap[1]:.1f}mm", y=1.02)
    plt.show()
```

```python
def plot_rain_regimes_hexbin(data_dict, models, crit='all',
                             constrain_var='prw', 
                             x_var='omega850', 
                             target_var='total_total_rain',
                             gridsize=25, # Controls resolution of hexagons
                             cmap='inferno_r'): # 'inferno_r' goes from light (low count) to dark (high count)
    
    # 1. Get Constraints (PRW Overlap)
    overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=constrain_var)
    if not overlap: return
    
    print(f"Constraining {constrain_var} to overlap: {overlap[0]:.2f} - {overlap[1]:.2f}")

    # 2. Determine Global Axis Limits (Extent) for consistent hexbins
    # We scan all models to find the min/max X (Omega) and Y (PRW) to lock the grid
    x_all, y_all = [], []
    
    # Temporary storage to avoid reloading data twice
    loaded_data = {} 

    # Load Dataframes
    for model in models:
        try:
            df_c = data_dict[model][crit][constrain_var]
            df_x = data_dict[model][crit][x_var]
            
            
            # Filter Time/Radius
            f_c = df_c[(df_c.radius==2) & (df_c.time_offset_hours==0)]
            f_x = df_x[(df_x.radius==2) & (df_x.time_offset_hours==0)]
            
            # Merge
            merged = pd.merge(f_c[['track_id','mean',target_var]], f_x[['track_id','mean',target_var]], 
                          on=['track_id',target_var], suffixes=('_c', '_x'))
            # merged = pd.merge(m1, df_t[['track_id', target_var]], on='track_id')
            
            # Filter by Overlap
            data = merged[(merged['mean_c'] >= overlap[0]) & (merged['mean_c'] <= overlap[1])].copy()
            
            if not data.empty:
                loaded_data[model] = data
                x_all.extend(data['mean_x'].values)
                y_all.extend(data['mean_c'].values)
        except: continue
            
    if not loaded_data:
        print("No data found.")
        return

    # Define the Global Extent [xmin, xmax, ymin, ymax]
    # We add a tiny buffer so points on the edge aren't cut off
    extent = [np.min(x_all), np.max(x_all), np.min(y_all), np.max(y_all)]

    # 3. Plotting Setup
    rows = len(models)
    cols = 2
    fig, axes = plt.subplots(rows, cols, figsize=(6, 2*rows),  #figsize=(10, 3.5*rows), 
                             sharex=True, sharey=True, constrained_layout=True)
    
    # If only 1 model, axes is 1D array, make it 2D for consistency
    if rows == 1: axes = np.array([axes])

    for idx, model in enumerate(models):
        if model not in loaded_data:
            axes[idx, 0].text(0.5, 0.5, "No Data", ha='center')
            continue
            
        df = loaded_data[model]
        
        # Calculate Split Threshold (Median P50)
        # You can change this to a fixed value (e.g., 5mm) if you prefer absolute comparisons
        # threshold = df[target_var].median()
        threshold_l = df[target_var].quantile(0.25); threshold_h = df[target_var].quantile(0.75)
        
        # Split Data
        df_low = df[df[target_var] < threshold_l] #df_low = df[df[target_var] <= threshold]
        df_high = df[df[target_var] > threshold_h]
        
        # --- Column 0: Low Intensity ---
        ax_low = axes[idx, 0]
        hb_low = ax_low.hexbin(df_low['mean_x'], df_low['mean_c'], 
                               gridsize=gridsize, extent=extent,
                               cmap=cmap, mincnt=1, linewidths=0.2)
        
        # Annotations
        count_low = len(df_low)
        ax_low.text(0.05, 0.95, f"Low Rain \n(<{threshold_l:.1f}mm)\nN={count_low}", 
                    transform=ax_low.transAxes, va='top', fontsize=9, 
                    bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))
        
        # --- Column 1: High Intensity ---
        ax_high = axes[idx, 1]
        hb_high = ax_high.hexbin(df_high['mean_x'], df_high['mean_c'], 
                                 gridsize=gridsize, extent=extent,
                                 cmap=cmap, mincnt=1, linewidths=0.2)
        
        count_high = len(df_high)
        ax_high.text(0.05, 0.95, f"High Rain \n(>{threshold_h:.1f}mm)\nN={count_high}", 
                     transform=ax_high.transAxes, va='top', fontsize=9,
                     bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))

        # --- Row Labeling ---
        # Add Model Name to the left of the row
        ax_low.set_ylabel(f"{MODEL_STYLES[model]['label']}\n{constrain_var}", fontsize=11, fontweight='bold')
        
        # Add colorbars for density (optional, but helpful per subplot)
        # Using inset_axes to place it inside or tiny next to plot
        # For cleanliness, we rely on the density color relative to itself
        
    # 4. Final Formatting
    # Set X-labels only on bottom row
    for ax in axes[-1, :]:
        ax.set_xlabel(x_var)

    # Set Column Titles
    axes[0, 0].set_title(f"Less Intense Precip (< P25)", fontsize=10.5)
    axes[0, 1].set_title(f"Most Intense Precip (> P75)", fontsize=10.5)
    
    plt.suptitle(f"Phase Space Density: {constrain_var} vs {x_var}(Constrained {constrain_var}: {overlap[0]:.1f}-{overlap[1]:.1f})", y=1.03, fontsize=10)
    plt.show()

# Usage:
# plot_rain_regimes_hexbin(processed_data, ['scream', 'um', 'icon'], crit='min2_longremoved')
```

```python
def plot_normalized_rain_omega_hexbin(data_dict, models, crit='all',
                                      constrain_var='prw', manual_prw_limits=(52, 53),
                                      x_var='omega850', 
                                      y_var='total_total_rain',
                                      gridsize=20, 
                                      cmap='Blues'):
    
    print(f"Applying Manual Constraint: {constrain_var} between {manual_prw_limits[0]} and {manual_prw_limits[1]}")

    # Setup Plot Grid
    n_models = len(models)
    cols = 3
    rows = (n_models // cols) + (1 if n_models % cols > 0 else 0)
    
    fig, axes = plt.subplots(rows, cols, figsize=(3*cols, 3*rows), constrained_layout=True)
    axes = axes.flatten()

    for idx, model in enumerate(models):
        ax = axes[idx]
        
        try:
            # 1. Load FULL Data (to calculate Global Min/Max for Normalization)
            # We need the full distribution to know what "0" and "1" mean for this model
            df_c_full = data_dict[model][crit][constrain_var]
            df_x_full = data_dict[model][crit][x_var]
            # df_y_full = data_dict[model][crit][y_var] # Rain
            
            if df_x_full is None: continue

            # Calculate Global Min/Max for Normalization (Per Model)
            # Using 1st and 99th percentile is often safer than absolute min/max to avoid outliers,
            # but here I use absolute min/max as requested.
            x_min, x_max = df_x_full['mean'].min(), df_x_full['mean'].max()
            y_min, y_max = df_c_full[y_var].quantile(0.05) , df_c_full[y_var].quantile(0.95) # 
            
            # 2. Filter Data based on Manual Constraint
            # Filter Time/Radius first
            f_c = df_c_full[(df_c_full.radius==2) & (df_c_full.time_offset_hours==0)]
            f_x = df_x_full[(df_x_full.radius==2) & (df_x_full.time_offset_hours==0)]
            
            # Merge
            merged = pd.merge(f_c[['track_id','mean',y_var]], f_x[['track_id','mean',y_var]], 
                          on=['track_id',y_var], suffixes=('_c', '_x'))
            # merged = pd.merge(m1, df_y_full[['track_id', y_var]], on='track_id')
            
            # Apply Manual PRW Constraint
            subset = merged[
                (merged['mean_c'] >= manual_prw_limits[0]) & 
                (merged['mean_c'] <= manual_prw_limits[1])
            ].copy()
            
            if subset.empty:
                ax.text(0.5, 0.5, "No Data in Range", ha='center', va='center')
                continue
                
            # 3. Normalize the SUBSET using the GLOBAL limits
            # This places the specific PRW slice into the context of the whole model
            subset['norm_x'] = (subset['mean_x'] - x_min) / (x_max - x_min)
            subset['norm_y'] = (subset[y_var] - y_min) / (y_max - y_min)
            
            # Clip to 0-1 just in case
            subset['norm_x'] = subset['norm_x'].clip(0, 1)
            subset['norm_y'] = subset['norm_y'].clip(0, 1)

            # 4. Plot Hexbin
            # X = Normalized Omega, Y = Normalized Rain, C = Density (default)
            hb = ax.hexbin(subset['norm_x'], subset['norm_y'], 
                           gridsize=gridsize, 
                           cmap=cmap, 
                           extent=[0, 1, 0, 1], # Force plotting area to normalized unit square
                           mincnt=1, 
                           linewidths=0.2)
            
            # 5. Annotations and Style
            ax.set_title(f"{MODEL_STYLES[model]['label']}\n(N={len(subset)})")
            
            # Add physical values labels to corners so we know what 0 and 1 mean
            ax.text(0.02, -0.02, f"X_min:\n{x_min:.2f}", transform=ax.transAxes, 
                    fontsize=7, va='top', ha='left', color='gray')
            ax.text(0.98, -0.02, f"X_max:\n{x_max:.2f}", transform=ax.transAxes, 
                    fontsize=7, va='top', ha='right', color='gray')
            ax.text(-0.02, 0.98, f"Y_max:\n{y_max:.1f}", transform=ax.transAxes, 
                    fontsize=7, va='top', ha='right', rotation=90, color='gray')

            ax.set_xlim(0, 1)
            ax.set_ylim(0, 1)
            
            # Only put axis labels on the edges
            if idx >= len(models) - cols:
                ax.set_xlabel(f"Norm. {x_var}")
            if idx % cols == 0:
                ax.set_ylabel(f"Norm. {y_var}")

        except Exception as e:
            print(f"Error {model}: {e}")

    # Hide unused axes
    for i in range(len(models), len(axes)):
        axes[i].axis('off')

    plt.suptitle(f"Norm. Rain vs Omega | Fixed {constrain_var}: {manual_prw_limits[0]}-{manual_prw_limits[1]}", y=1.02)
    plt.show()

# --- Usage Example ---
# plot_normalized_rain_omega_hexbin(processed_data, 
#                                   ['scream', 'um', 'icon', 'ifs', 'nicam', 'obs'], 
#                                   manual_prw_limits=(52, 53),
#                                   crit='min2_longremoved')
```

```python
def plot_mean_rain_vs_binned_omega(data_dict, models, crit='all',
                                   constrain_var='prw', manual_prw_limits=(52, 53),
                                   omega_var='omega850', rain_var='total_total_rain',
                                   n_omega_bins=10, stat='mean', radius=2, time=0):
    
    # 1. Setup Data Collection and Global Binning
    processed_dfs = []
    
    # Collect all Omega values within the PRW constraint to define a COMMON X-axis for all models
    all_omega_vals = []

    for model in models:
        try:
            df_c = data_dict[model][crit][constrain_var]
            df_x = data_dict[model][crit][omega_var]
            
            # Filter Time/Radius
            f_c = df_c[(df_c.radius==radius) & (df_c.time_offset_hours==time)]
            f_x = df_x[(df_x.radius==radius) & (df_x.time_offset_hours==time)]
            
            # Merge and Filter
            merged = pd.merge(f_c[['track_id','mean',rain_var]], f_x[['track_id','mean',rain_var]], 
                              on=['track_id',rain_var], suffixes=('_c', '_x'))
            
            # Apply Manual PRW Constraint
            subset = merged[
                (merged['mean_c'] >= manual_prw_limits[0]) & 
                (merged['mean_c'] <= manual_prw_limits[1])
            ].copy()
            
            if not subset.empty:
                all_omega_vals.extend(subset['mean_x'].values)
        except Exception as e:
            # print(f"Skipping model {model}: {e}")
            continue

    if not all_omega_vals:
        print("No data found in the specified PRW range.")
        return

    # Define COMMON Omega Bins (X-axis) across all models 
    # Use 1st and 99th percentile for robust bin edges
    omega_edges = np.linspace(
        np.percentile(all_omega_vals, 1), 
        np.percentile(all_omega_vals, 99), 
        n_omega_bins + 1
    )
    
    # 2. Process Data Per Model
    for model in models:
        try:
            # Load and Filtered data (same steps as above, but now merge rain)
            df_c = data_dict[model][crit][constrain_var]
            df_x = data_dict[model][crit][omega_var]
            # df_y = data_dict[model][crit][rain_var] # Rain
            
            f_c = df_c[(df_c.radius==radius) & (df_c.time_offset_hours==time)]
            f_x = df_x[(df_x.radius==radius) & (df_x.time_offset_hours==time)]
            
            # Merge all 
            full_merged = pd.merge(f_c[['track_id','mean',rain_var]],
                                   f_x[['track_id','mean',rain_var]], on=['track_id',rain_var],
                                   suffixes=('_c', '_x'))

            # Apply Manual PRW Constraint
            subset = full_merged[
                (full_merged['mean_c'] >= manual_prw_limits[0]) & 
                (full_merged['mean_c'] <= manual_prw_limits[1])
            ].copy()
            
            if subset.empty: continue
        
            # This creates a Series with the bins as the index
            omega_bins = pd.cut(subset['mean_x'], bins=omega_edges, include_lowest=True)
            
            # Group by bin and calculate the mean/median rain rate
            # The result has the pandas.IntervalIndex
            agg_results = subset.groupby(omega_bins, observed=False)[rain_var].agg([stat, 'count'])
            
            # 1. Calculate the omega_center (the numerical X-coordinate for plotting)
            # Use the IntervalIndex directly for the calculation
            agg_results['omega_center'] = agg_results.index.map(lambda x: x.mid).astype(float)
            
            # 2. Reset the index and DUMP the 'omega_bin' interval column
            # This uses the original RangeIndex (0, 1, 2...) which is safe for concatenation
            agg_results = agg_results.reset_index(drop=True)
            
            # 3. Add metadata columns
            agg_results['Model'] = model
            agg_results['Model_Label'] = MODEL_STYLES[model]['label']
            
            processed_dfs.append(agg_results)
            
        except Exception as e:
            # print(f"Error processing {model}: {e}")
            continue

    if not processed_dfs:
        print("No models had sufficient data in the constrained range.")
        return

    final_df = pd.concat(processed_dfs)
    # We map the 'Model_Label' back to the color defined in MODEL_STYLES[model]
    unique_labels = final_df[['Model', 'Model_Label']].drop_duplicates()

    # 3. Plot
    palette = {row.Model_Label: MODEL_STYLES[row.Model]['color'] for _, row in unique_labels.iterrows()}


    # 3. Plotting
    fig, ax = plt.subplots(figsize=(10, 6))

    # Plot the mean/median rain rate
    sns.lineplot(
        data=final_df, 
        x='omega_center', 
        y=stat, 
        hue='Model_Label', 
        palette = {row.Model_Label: MODEL_STYLES[row.Model]['color'] for _, row in unique_labels.iterrows()},
        # palette=palette, #{m: MODEL_STYLES[m]['color'] for m in models},
        ax=ax, 
        linewidth=2.5
    )
    
    # Add a vertical line at omega=0 (neutral buoyancy)
    ax.axvline(0, color='gray', linestyle='--', alpha=0.7)

    ax.set_xlabel(f"Binned {omega_var} (Pa/s)", fontsize=12)
    ax.set_ylabel(f"{stat.capitalize()} Rain Rate (mm/hr)", fontsize=12)
    ax.set_title(f"{stat.capitalize()} Rain Rate vs. Dynamics\nConstrained {constrain_var}: {manual_prw_limits[0]:.2f}-{manual_prw_limits[1]:.2f}", fontsize=14)

    # Improve Legend
    sns.move_legend(ax, "upper left", title="Model", frameon=False)
    ax.grid(True, linestyle=':', alpha=0.6)
    
    # 4. (Optional) Show N count
    # You can optionally create a twin axis or a second plot to show the number of cases (count) per bin, 
    # as reliability decreases where N is small.
    
    plt.show()
```

```python
def plot_rain_vs_omega_constrained(data_dict, models, crit='all', 
                                   constrain_var='prw', 
                                   x_var='omega850', 
                                   y_var='total_total_rain',  radius=2, time=0):
    
    # 1. Get constraint range (Intersection of 10th-90th percentiles)
    overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=constrain_var)
    if not overlap: return

    plot_data = []

    for model in models:
        # We need to merge the Constrain Var DF and the X Var DF
        # Assuming they share 'track_id', 'radius', 'time_offset_hours'
        
        df_constrain = data_dict[model][crit][constrain_var]
        df_x = data_dict[model][crit][x_var]
        
        # Merge on index or common columns. 
        # Crucial: Ensure 'radius' and 'time' are filtered BEFORE merge to avoid duplicates
        # or merge on them too.
        
        # Simple approach: Filter both first
        f_c = df_constrain[(df_constrain.radius==radius) & (df_constrain.time_offset_hours==time)]
        f_x = df_x[(df_x.radius==radius) & (df_x.time_offset_hours==time)]
        
        # Merge. We need 'track_id' to match the exact same storm event
        merged = pd.merge(f_c[['track_id', 'mean', y_var]], 
                          f_x[['track_id', 'mean']], 
                          on='track_id', suffixes=('_constrain', '_x'))
        
        # Filter by Constraint (e.g., PRW between 50 and 55)
        filtered = merged[
            (merged['mean_constrain'] >= overlap[0]) & 
            (merged['mean_constrain'] <= overlap[1])
        ].copy()
        
        filtered['Model'] = MODEL_STYLES[model]['label']
        plot_data.append(filtered)

    full_df = pd.concat(plot_data)

    # 2. Bin the X-variable (Omega)
    # create 10 bins for Omega
    full_df['omega_bin'] = pd.cut(full_df['mean_x'], bins=10)

    # 3. Plot
    plt.figure(figsize=(12, 6))
    
    # Boxplot: X=Omega Bin, Y=Rain, Hue=Model
    sns.boxplot(data=full_df, x='omega_bin', y=y_var, hue='Model', 
                palette={v['label']: v['color'] for k,v in MODEL_STYLES.items() if k in models},
                showfliers=False) # Hide outliers to see the distribution box better
    
    plt.title(f"Rain vs Omega850 (Constrained: {overlap[0]:.1f} < {constrain_var} < {overlap[1]:.1f})")
    plt.xticks(rotation=45)
    plt.xlabel("Omega 850hPa Bins (Pa/s)")
    plt.ylabel("Total Rainfall Distribution")
    plt.show()
```

### Correlation Matrix Analysis

Analyze correlations between environmental variables within a specific time window

#### Function: `plot_correlation_matrix`

This function computes and visualizes correlations between environmental variables within a specified time window.

**Key Features:**
- Filter data by time window (e.g., pre-convective: -6 to 0 hours)
- Filter by radius
- Aggregate values across the time window for each track
- Compute Pearson correlation coefficients
- **NEW:** Include MCS-specific statistics (e.g., total_total_rain, duration_hours, total_area) in correlation analysis
- Optional masking of upper triangle (to avoid redundancy)
- Customizable appearance (colormap, annotations, size)

**Parameters:**
- `var_dfs_dict`: Dictionary of variable DataFrames (e.g., `var_dfs_scream`)
- `time_window`: Tuple of (start_hour, end_hour) relative to MCS initiation
- `radius`: Radius in degrees to analyze (default: 2.0)
- `mcs_stats_df`: (Optional) DataFrame with MCS statistics (e.g., `df_trop_scream_all`)
- `mcs_stats_cols`: (Optional) List of column names from mcs_stats_df to include (e.g., `['total_total_rain']`)
- `mask_upper`: Whether to mask upper triangle (default: True)
- `figsize`, `cmap`, `annot`, `fontsize`: Visualization options

**Returns:**
- `fig`: Matplotlib figure object
- `corr_matrix`: pandas DataFrame with correlation coefficients
