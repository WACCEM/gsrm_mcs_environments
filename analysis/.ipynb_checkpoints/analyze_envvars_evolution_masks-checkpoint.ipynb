```python
main_dir = '/pscratch/sd/p/paccini/temp/hackathon/updated_environmental_variables/'
mcs_stats = 'df_filtered_mcs_track_ids/'
# main_dir = '/Users/pacc275/local_documents/hackathon/updated_environmental_variables/' #local
# --- Define main components ---
TRACK_MODELS = ['um', 'icon', 'scream', 'ifs', 'nicam', 'obs', 'obs_v7'] 
# VARIABLES = ['prw', 'rh500', 'rh850', 'q500', 'q850', 'omega500', 'omega850',  'lShear', 'dShear','free_rh'] #'w850', 'w500',
# VARIABLES = ['prw',  'q850', 'omega850',  'lShear', 'dShear','free_rh']
VARIABLES = ['prw',  'q850', 'omega850',  'lShear', 'dShear','free_rh']
# VARIABLES = ['prw', 'omega850', ] #'w850', 'w500',
CRITERIA = ['all', 'min2']
```

```python
# --- Path Configuration ---

# 1. Map your track model name (e.g., 'obs') to the 
#    corresponding environmental data directory (e.g., 'era5')
ENV_MODEL_DIR_MAP = {
    'um': 'UM_all/from_masks/',
    'icon': 'ICON_all/from_masks/',
    'scream': 'SCREAM_all/from_masks/',
    'ifs': 'IFS_all/from_masks/',
    'nicam': 'NICAM_all/from_masks/',
    'obs_v7': 'ERA5_IMERGv7_all/from_masks/' ,
    'obs': 'era5/from_masks/'  # Key insight: 'obs' tracks map to 'era5' env data
}

# 2. Define the path pattern for each variable.
#    'type': 'dir' (read the whole dir as parquet) or 
#            'glob' (find a single file using a wildcard '*')
VAR_PATH_PATTERNS = {
    # Default patterns used by most models (um, scream, etc.)
    'default': {
        'prw':      {'type': 'dir', 'pattern': 'prw/'},
        'rh500':    {'type': 'glob', 'pattern': 'hur/mcs_env_hur_*_hur_500hPa.parquet'},
        'rh850':    {'type': 'glob', 'pattern': 'hur/mcs_env_hur_*_hur_850hPa.parquet'},
        'free_rh':    {'type': 'glob', 'pattern': 'hur/mcs_env_hur_*_hur_avg850-*-500hPa.parquet'},
        'q500':     {'type': 'glob', 'pattern': 'hus/mcs_env_hus_*_hus_500hPa.parquet'},
        'q850':     {'type': 'glob', 'pattern': 'hus/mcs_env_hus_*_hus_850hPa.parquet'},
        'omega500': {'type': 'glob', 'pattern': 'omega/mcs_env_omega_*_omega_500hPa.parquet'},
        'omega850': {'type': 'glob', 'pattern': 'omega/mcs_env_omega_*_omega_850hPa.parquet'},
        # 'w500':     {'type': 'glob', 'pattern': 'wa/mcs_env_wa_*_wa_500hPa.parquet'},
        # 'w850':     {'type': 'glob', 'pattern': 'wa/mcs_env_wa_*_wa_850hPa.parquet'},
        'lShear':   {'type': 'dir', 'pattern': 'low_shear_magnitude_975hPa-800hPa/'},
        'dShear':   {'type': 'dir', 'pattern': 'deep_shear_magnitude_850hPa-400hPa/'},
    },
    # Overrides for the 'obs' (era5) model
    'obs': {
        'q500':     {'type': 'dir', 'pattern': 'q_500mb/'},
         'rh500':  {'type': 'dir', 'pattern': 'rh_500mb/'},
        'q850':     {'type': 'dir', 'pattern': 'q_850mb/'},
         'rh850':  {'type': 'dir', 'pattern': 'rh_850mb/'},
          'prw': {'type': 'dir', 'pattern':'TCWV/'},
        'rh500':  {'type': 'dir', 'pattern': 'rh_500mb/'},
        'omega500':     {'type': 'dir', 'pattern': 'w_500mb/'},
        'omega850':     {'type': 'dir', 'pattern': 'w_850mb/'},
        # 'lShear':   {'type': 'dir', 'pattern': 'shear_mag_0to2km/'},
        # 'dShear':   {'type': 'dir', 'pattern': 'shear_mag_0to10km/'},
    }
}

# --- New Path-Finding Function ---

def get_env_path(base_out_dir, model_name, var_name):
    """
    Finds the correct file path for a given model and variable,
    handling inconsistent names and date stamps.
    
    'model_name' should be from TRACK_MODELS (e.g., 'um', 'obs')
    """
    
    # 1. Get the top-level directory name (e.g., 'UM_all' or 'era5')
    if model_name not in ENV_MODEL_DIR_MAP:
        raise KeyError(f"Model '{model_name}' not defined in ENV_MODEL_DIR_MAP.")
    model_dir = ENV_MODEL_DIR_MAP[model_name]
    
    # 2. Get the path pattern rule for this variable
    # Check for a model-specific override first (e.g., for 'obs')
    if model_name in VAR_PATH_PATTERNS and var_name in VAR_PATH_PATTERNS[model_name]:
        rule = VAR_PATH_PATTERNS[model_name][var_name]
    # Otherwise, use the default rule
    elif var_name in VAR_PATH_PATTERNS['default']:
        rule = VAR_PATH_PATTERNS['default'][var_name]
    else:
        raise KeyError(f"Variable '{var_name}' not defined in VAR_PATH_PATTERNS.")
        
    pattern_type = rule['type']
    pattern = rule['pattern']
    
    # 3. Construct the full search path
    full_search_path = os.path.join(base_out_dir, model_dir, pattern)
    
    # 4. Find the file/directory
    if pattern_type == 'dir':
        # For 'dir' type, we just return the path as-is.
        return full_search_path
        
    elif pattern_type == 'glob':
        # For 'glob', we find all files matching the pattern
        search_results = glob.glob(full_search_path)
        
        if len(search_results) == 0:
            raise FileNotFoundError(f"No file found for {model_name}/{var_name}. "
                                    f"Searched: {full_search_path}")
        if len(search_results) > 1:
            # This is ambiguous, which is bad.
            raise FileNotFoundError(f"Multiple files found for {model_name}/{var_name}. "
                                     f"Searched: {full_search_path}. "
                                     f"Found: {search_results}")
        
        # Success! Return the one and only match.
        return search_results[0]
```

```python
# --- Transformation Configuration ---
# Define all model/variable-specific unit conversions here

TRANSFORM_CONFIG = {
    # 'scream': {
    #     'rh500': lambda df: df.assign(mean=df['mean'] * 100),
    #     'rh850': lambda df: df.assign(mean=df['mean'] * 100),
    # },
    'nicam': {
        'rh500': lambda df: df.assign(mean=df['mean'] * 100),
        'rh850': lambda df: df.assign(mean=df['mean'] * 100),
        'free_rh': lambda df: df.assign(mean=df['mean'] * 100),
    },
    'obs': {  # 'obs' is our alias for 'era5imergv6'
        'q500': lambda df: df.assign(mean=df['mean'] / 1000),
        'q850': lambda df: df.assign(mean=df['mean'] / 1000),
    },

}
```

```python
# --- Load Environmental Data ---
raw_env_data = {}
for model in TRACK_MODELS: # Loop over our base track models
    raw_env_data[model] = {}
    for var in VARIABLES:
        try:
            # 1. Find the path (same as before)
            path = get_env_path(main_dir, model, var)
            
            # 2. Load the data
            df = pd.read_parquet(path)
            
            # 3. *** NEW: Apply Transformation ***
            # Get the specific transform function from our config
            # If no rule exists, it defaults to a function that does nothing (lambda x: x)
            transform_func = TRANSFORM_CONFIG.get(model, {}).get(var, lambda x: x)
            
            # Apply the function to the DataFrame
            df_transformed = transform_func(df)
            
            # 4. Store the standardized DataFrame
            raw_env_data[model][var] = df_transformed
        #     # Our function handles all the ugly logic!
        #     path = get_env_path(main_dir, model, var)
        #     raw_env_data[model][var] = pd.read_parquet(path)
        except Exception as e:
            # Catch errors from loading OR transforming
            print(f"Warning: Could not load env data for {model} - {var}. Error: {e}")
            raw_env_data[model][var] = None

# --- Load Track Data ---
raw_track_data = {}
for model in TRACK_MODELS:
    raw_track_data[model] = {}
    for crit in CRITERIA:
        try:
            crit_suffix = f'_{crit}' if crit != 'all' else ''
            # path = os.path.join(mcs_stats, f'df_trop_{model}{crit_suffix}.pkl')
            path = os.path.join(mcs_stats, f'with_circular_lffiltered_{model}{crit_suffix}_r2.pkl')
            raw_track_data[model][crit] = pd.read_pickle(path)
        except Exception as e:
            print(f"Warning: Could not load track data for {model} - {crit}. Error: {e}")
            raw_track_data[model][crit] = None
```

```python
# --- Filtering environments ---
# def get_filtered_df(df_model, df_tracks, model_track='track_id', filter_track='track_id'):
#     if df_model is None or df_tracks is None:
#         return None
#     filtered_df = df_model[df_model[model_track].isin(df_tracks[filter_track])]
#     return filtered_df
def get_filtered_df(df_main, df_tracks, lf_thresh=0.1,lf_var='total_land_fraction_track'):
    """
    Merges main data with track data data and filters by circular land fraction threshold.
    """
    # Safe Merge (Inner Join) on track_id to ensure data integrity
    # Adjust 'track_id' if your column names differ between dataframes
    merged_df = pd.merge( df_main, df_tracks[['track_id', 'total_total_rain',lf_var]], on='track_id', how='inner')
    
    # Filter by Land Fraction Threshold
    # Ensure it handles NaNs if necessary, though inner join usually drops non-matches
    final_df = merged_df[merged_df[lf_var] <= lf_thresh].copy()
    
    return final_df

def remove_longdurations(df_model, feature='duration_hours', p_high=0.9):
    if df_model is None:
        return None
    q_high = df_model[feature].quantile(p_high)
    df_filtered = df_model[(df_model[feature] <= q_high)].copy()
    return df_filtered
```

```python
remove_longdurations(raw_track_data['obs']['min2'])
```

```python
# --- Final processing loop  ---
processed_data = {}

for model in TRACK_MODELS:
    processed_data[model] = {}
    for crit in CRITERIA:
        base_track_df = raw_track_data[model].get(crit)
        
        if base_track_df is None:
            print(f"Skipping {model} - {crit}: Missing base track data.")
            continue
            
        long_removed_track_df = remove_longdurations(base_track_df)
        
        crit_key_base = crit
        crit_key_longremoved = f'{crit}_longremoved'
        
        processed_data[model][crit_key_base] = {}
        processed_data[model][crit_key_longremoved] = {}
        
        for var in VARIABLES:
            raw_env_df = raw_env_data[model].get(var)
            
            if raw_env_df is None:
                # This will now correctly skip missing 'era5' data
                continue 

            processed_data[model][crit_key_base][var] = get_filtered_df(
                raw_env_df, base_track_df
            )
            
            processed_data[model][crit_key_longremoved][var] = get_filtered_df(
                raw_env_df, long_removed_track_df
            )

print("Processing complete!")
```

### Choosing similar environments for conditional probability distributions

```python
## 1. Objectively calculate overlapping environmental conditions
def get_common_overlap(data_dict, models, crit='all', var='prw', low_p=10, high_p=90):
    """
    Calculates the intersection of ranges [P_low, P_high] across all models.
    """
    global_min = -np.inf
    global_max = np.inf
    
    ranges = {}
    
    for model in models:
        # --- FIX: Access dictionary dynamically using 'crit' ---
        try:
            df = data_dict[model][crit][var]
        except KeyError:
            print(f"Warning: Data not found for {model} -> {crit} -> {var}")
            return None, None

        if df is None:
            print(f"Warning: DataFrame is None for {model} -> {crit} -> {var}")
            return None, None
        
        # Extract values for the CONTROL variable (e.g., PRW)
        values = df['mean'].values
        
        # Calculate percentiles
        lower = np.percentile(values, low_p)
        upper = np.percentile(values, high_p)
        ranges[model] = (lower, upper)
        
        # Update global intersection
        global_min = max(global_min, lower)
        global_max = min(global_max, upper)
    
    # Check if a valid overlap exists
    if global_min >= global_max:
        print(f"No overlap found! Min of tops ({global_max:.2f}) < Max of bottoms ({global_min:.2f})")
        return None, ranges
        
    return (global_min, global_max), ranges
```

```python
### 2. Total precipitation amount distribution in the "common-range" environmental conditions
def plot_rain_in_common_env(data_dict, models, crit='all', 
                            control_var='prw', overlap=None,
                            target_var='total_total_rain', 
                            radius=2, time=0):
    if overlap==None:
    # 1. Get Objective Range using the corrected function
        overlap, ranges = get_common_overlap(data_dict, models, crit=crit, var=control_var)
    
    # if not overlap:
    #     return
    
    print(f"Comparison Criteria: {crit}")
    print(f"Common {control_var} range: {overlap[0]:.2f} - {overlap[1]:.2f}")

    # 2. Collect Data
    plot_data = []
    
    for model in models:
        # --- FIX: Access dictionary dynamically ---
        try:
            df = data_dict[model][crit][control_var]
        except KeyError:
            continue

        if df is None: continue
        
        # Apply Filters:
        # 1. Radius & Time
        # 2. Control Variable Range (e.g. PRW is within the overlap)
        mask = (
            (df['radius'] == radius) & 
            (df['time_offset_hours'] == time) & 
            (df['mean'] >= overlap[0]) & 
            (df['mean'] <= overlap[1])
        )
        
        subset = df[mask].copy()
        
        if subset.empty:
            print(f"Warning: {model} has no data in the common range.")
            continue
        
        # Keep only the target rain data and label it
        subset = subset[[target_var]].rename(columns={target_var: 'rain'})
        subset['Model'] = MODEL_STYLES[model]['label'] 
        plot_data.append(subset)
    
    if not plot_data:
        print("No data available to plot.")
        return

    full_df = pd.concat(plot_data)
    
    # 3. Plot
    plt.figure(figsize=(7, 4))
    
    # KDE Plot
    sns.kdeplot(data=full_df, x='rain', hue='Model', 
                palette={v['label']: v['color'] for k,v in MODEL_STYLES.items() if k in models},
                linewidth=2.5, common_norm=False)
    # sns.move_legend(title=None, fontsize=8)
    plt.title(f"Rain Dist. ({crit})\nwhere {overlap[0]:.1f} < {control_var} < {overlap[1]:.1f}")
    plt.xlabel("Total Rainfall (mm)"); plt.xscale('log')
    plt.xlim(left=0) # Lock left axis to 0 for rain
    plt.grid(True, alpha=0.3)
    plt.show()

# --- RUN CONFIGURATION ---
# Case: Similar Distributions (Excluding SCREAM/UM outliers for PRW)
# p1. Define the models you want to compare
# selected_models = ['icon', 'ifs', 'obs', 'nicam'] 

# # 2. Run the plot with your specific criteria
# plot_rain_in_common_env(
#     processed_data, 
#     selected_models, 
#     crit='min2_longremoved',  # <--- Now handled correctly!
#     control_var='prw',
#     target_var='total_total_rain'
# )
```

```python
def plot_low_high_regime(data_dict, models, crit='all', 
                            control_var='prw', target_var='total_total_rain',
                            radius=2, time=0):
    
    # 1. Get Common Overlap Range
    overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=control_var)
    if not overlap: return
    
    print(f"Common {control_var} range: {overlap[0]:.2f} - {overlap[1]:.2f}")

    # 2. Collect Data & Calculate Shared Median
    # We need to collect all valid data first to find the median of the OVERLAP region
    all_valid_values = []
    model_data_store = {}

    for model in models:
        try:
            df = data_dict[model][crit][control_var]
        except KeyError: continue
        if df is None: continue

        # Filter: Radius, Time, and Overlap Range
        mask = (
            (df.radius == radius) & 
            (df.time_offset_hours == time) & 
            (df['mean'] >= overlap[0]) & 
            (df['mean'] <= overlap[1])
        )
        valid_data = df[mask].copy()
        model_data_store[model] = valid_data
        all_valid_values.extend(valid_data['mean'].values)

    if not all_valid_values:
        print("No data found in overlap region.")
        return

    # Calculate split point (median of the constrained environment)
    split_val = np.median(all_valid_values)
    print(f"Splitting Regimes at {control_var} = {split_val:.2f}")

    # 3. Plotting
    fig, axes = plt.subplots(1, 2, figsize=(9, 4), sharey=False, sharex=True)
    titles = [f"Low {control_var} (<{split_val:.1f})", f"High {control_var} (>{split_val:.1f})"]

    for i, ax in enumerate(axes):
        plot_dfs = []
        
        for model in models:
            if model not in model_data_store: continue
            
            df = model_data_store[model]
            
            # Filter Low vs High
            if i == 0: 
                subset = df[df['mean'] <= split_val]
            else:      
                subset = df[df['mean'] > split_val]
            
            # Create Label with Count (N)
            count = len(subset); 
            label_str = f"{MODEL_STYLES[model]['label']} \n(N={count})"
            
            # Prepare for seaborn
            temp = subset[[target_var]].copy()
            temp.columns = ['rain']
            temp['Model_Label'] = label_str
            temp['Model_Key'] = model # Keep key for color mapping
            plot_dfs.append(temp)
        
        if not plot_dfs: continue
        full_df = pd.concat(plot_dfs)
        
        # Create custom palette that maps the NEW labels to the OLD colors
        # We map the 'Model_Label' back to the color defined in MODEL_STYLES[model]
        unique_labels = full_df[['Model_Label', 'Model_Key']].drop_duplicates()

        # 3. Plot
        palette = {row.Model_Label: MODEL_STYLES[row.Model_Key]['color'] for _, row in unique_labels.iterrows()}
        # KDE Plot
        sns.kdeplot(data=full_df, x='rain', hue='Model_Label', 
                    palette=palette,ax=ax,
                    linewidth=2.5, common_norm=False)

        sns.move_legend(ax, "upper right", title=None, fontsize=8)
        # ax.legend(fontsize=10)
        ax.set_title(titles[i]); ax.set_xscale('log')
        ax.set_xlabel("Total Rain (mm)")
        
    plt.tight_layout()
    plt.show()
```

```python
def plot_rain_phase_space_normalized(data_dict, models, crit='all',
                                     constrain_var='prw', 
                                     x_var='omega850', 
                                     target_var='total_total_rain',n_y_bins=3,
                                     norm_method='quantile', # 'quantile' or 'minmax'
                                     p_low=0.05, p_high=0.95,cmap=mycmap.whiteTOblueTOgreen):
    
    # 1. Get Constraints and Global Ranges
    overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=constrain_var)
    if not overlap: return

    # To ensure bins are identical across subplots, we need Global Min/Max for the X-variable
    # We scan all models briefly to find the min/max of Omega within the PRW overlap
    x_vals_all = []
    y_vals_all = []
    for m in models:
        try:
            df_c = data_dict[m][crit][constrain_var]
            df_x = data_dict[m][crit][x_var]

            # Filter for common track_ids in the overlap range
            # (This assumes track_ids align, simplified for speed)
            valid_c = df_c[df_c['mean'].between(overlap[0], overlap[1])]
            valid_tracks = valid_c['track_id']
            
            # Store Y values (Constraint) strictly within overlap
            y_vals_all.extend(valid_c['mean'].values)
            
            # Rough filter to get relevant range
    
            relevant_x = df_x[df_x['track_id'].isin(valid_tracks)]
            x_vals_all.extend(relevant_x['mean'].values)
        except: continue
            
    if not x_vals_all or not y_vals_all: 
        print("No data found for binning.")
        return
    
    # Define Global Bins
    # Y-Bins: Percentiles of the data within the overlap
    # If n=2, quantiles=[0, 0.5, 1.0]. If n=3, quantiles=[0, 0.33, 0.66, 1.0]
    qs = np.linspace(0, 1, n_y_bins + 1)
    y_bins = np.unique(np.quantile(y_vals_all, qs)) # Unique handles cases with low precision

    # X-Bins: The Omega range divided into 5
    x_bins = np.linspace(np.percentile(x_vals_all, 1), np.percentile(x_vals_all, 99), 6) # 5 bins (cutting top/bottom 1% outliers)

    # 2. Prepare Subplots
    n_models = len(models)
    cols = 3
    rows = (n_models // cols) + (1 if n_models % cols > 0 else 0)
    fig, axes = plt.subplots(rows, cols, figsize=(5*cols, 4*rows), constrained_layout=True)
    axes = axes.flatten()

    mappable = None
    # 3. Loop Models
    for idx, model in enumerate(models):
        ax = axes[idx]
        
        try:
            # Load Dataframes
            df_c = data_dict[model][crit][constrain_var] # PRW
            df_x = data_dict[model][crit][x_var]         # Omega
            
            # Filter Radius/Time (Assumed 2/0 based on previous context, or pass as args)
            f_c = df_c[(df_c.radius==2) & (df_c.time_offset_hours==0)]
            f_x = df_x[(df_x.radius==2) & (df_x.time_offset_hours==0)]
            
            # Merge: C + X + T
            merged = pd.merge(f_c[['track_id','mean',target_var]], f_x[['track_id','mean',target_var]], on=['track_id',target_var], suffixes=('_c', '_x'))
            
            # Filter by Constraint Overlap
            data = merged[(merged['mean_c'] >= overlap[0]) & (merged['mean_c'] <= overlap[1])].copy()
            
            if data.empty:
                ax.text(0.5, 0.5, "No Overlap Data", ha='center', va='center')
                continue

            # Binning
            data['x_bin'] = pd.cut(data['mean_x'], bins=x_bins, include_lowest=True)
            data['y_bin'] = pd.cut(data['mean_c'], bins=y_bins, include_lowest=True)
            
            # Aggregate: Calculate Mean Rain per cell
            # We use unstack to create a matrix for the heatmap
            pivot_mean = data.groupby(['y_bin', 'x_bin'], observed=False)[target_var].mean().unstack()
            pivot_count = data.groupby(['y_bin', 'x_bin'], observed=False)['track_id'].count().unstack().fillna(0).astype(int)
            
            # --- NORMALIZATION (User Logic) ---
            # Flatten to apply normalization logic
            data_flat = pivot_mean.values
            
            if norm_method == 'quantile':
                vmin_model = np.nanquantile(data_flat, p_low)
                vmax_model = np.nanquantile(data_flat, p_high)
            else: # minmax
                vmin_model = np.nanmin(data_flat)
                vmax_model = np.nanmax(data_flat)
            
            # Normalize to 0-1 range
            denom = vmax_model - vmin_model
            if denom > 0:
                norm_data = (data_flat - vmin_model) / denom
                norm_data = np.clip(norm_data, 0, 1)
            else:
                norm_data = np.zeros_like(data_flat) # or 0.5 if flat
            
            # Custom colormap
            levels = MaxNLocator(nbins=20).tick_values(0, 1)
            norm = BoundaryNorm(levels, ncolors=plt.get_cmap(cmap).N, clip=True)
            
            # Reconstruct DataFrame for plotting
            norm_df = pd.DataFrame(norm_data, index=pivot_mean.index, columns=pivot_mean.columns)
            
            # Sort index descending so high PRW is at top of Y-axis
            norm_df = norm_df.sort_index(ascending=False) # High PRW on top

            # Apply the same sorting and shape to the Count matrix
            pivot_count_sorted = pivot_count.loc[norm_df.index, norm_df.columns]

            # Plot Heatmap
            im = sns.heatmap(norm_df, ax=ax, cmap=cmap, norm=norm,
                        cbar=False, cbar_kws={'label': 'Norm. Rain Intensity'},
                        vmin=0, vmax=1, annot=pivot_count_sorted, fmt="d", annot_kws={"size": 8})
            # Capture the mappable object from the Axes object
            if idx == 0:
                mappable = im.collections[0]
            
            ax.set_title(f"{MODEL_STYLES[model]['label']}\n(N={len(data)})")
            ax.set_xlabel(x_var)
            ax.set_ylabel(constrain_var)
            
            # Clean up tick labels for readability
            ax.set_xticklabels([f"{b.mid:.2f}" for b in norm_df.columns], rotation=45)
            ax.set_yticklabels([f"{b.mid:.1f}" for b in norm_df.index], rotation=0)

        except Exception as e:
            print(merged.columns)
            print(f"Error plotting {model}: {e}")
            ax.set_visible(False)

    # Turn off unused subplots
    for i in range(len(models), len(axes)):
        axes[i].axis('off')
    # Global Colorbar Application    
    if mappable:
        cbar_label = f"{'Normalized '}{target_var}"
        fig.colorbar(mappable, ax=axes.tolist(), orientation='vertical',label=cbar_label, 
                     shrink=0.8, pad=0.02, aspect=30)
            
        plt.suptitle(f"Norm. Rain vs {x_var} | Constrained {constrain_var}: {overlap[0]:.1f}-{overlap[1]:.1f}mm", y=1.02)
        plt.show()

# --- HOW TO RUN ---
# selected = ['scream', 'um', 'icon', 'ifs', 'nicam', 'obs']
# plot_rain_phase_space_normalized(processed_data, selected, crit='min2_longremoved')
```

```python
def plot_rain_phase_space_normalized(data_dict, models, crit='all',
                                     constrain_var='prw', 
                                     x_var='omega850', 
                                     target_var='total_total_rain', n_y_bins=3,
                                     n_x_bins=5, # New parameter for consistent X-bin count
                                     norm_method='quantile', 
                                     p_low=0.05, p_high=0.95, cmap=mycmap.whiteTOblueTOgreen): 
    
    # 1. Get Constraints and Global Ranges (PRW overlap)
    overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=constrain_var)
    if not overlap: return

    # Collect Y values only for GLOBAL Y-bin calculation
    y_vals_all = []
    for m in models:
        try:
            df_c = data_dict[m][crit][constrain_var]
            valid_c = df_c[df_c['mean'].between(overlap[0], overlap[1])]
            y_vals_all.extend(valid_c['mean'].values)
        except: continue
            
    if not y_vals_all: return
    
    # Define Global Y-Bins (PRW - based on percentiles, consistent across models)
    qs = np.linspace(0, 1, n_y_bins + 1)
    y_bins = np.unique(np.quantile(y_vals_all, qs))

    # 2. Prepare Subplots and Global Colorbar Setup
    n_models = len(models)
    cols = 3
    rows = (n_models // cols) + (1 if n_models % cols > 0 else 0)
    fig, axes = plt.subplots(rows, cols, figsize=(4*cols, 3*rows), sharey=True,constrained_layout=True)
    axes = axes.flatten()

    # Define global colorbar norm based on the final normalized range [0, 1]
    levels = MaxNLocator(nbins=20).tick_values(0, 1)
    # Ensure 'cmap' is a valid Matplotlib Colormap object or string
    cmap_obj = plt.get_cmap(cmap) if isinstance(cmap, str) else cmap 
    global_norm = BoundaryNorm(levels, ncolors=cmap_obj.N, clip=True)

    mappable = None
    
    # 3. Loop Models (Main Logic)
    for idx, model in enumerate(models):
        ax = axes[idx]
        
        try:
            # Load Dataframes
            df_c = data_dict[model][crit][constrain_var] 
            df_x = data_dict[model][crit][x_var]          
            
            # Filter Radius/Time & Merge
            f_c = df_c[(df_c.radius==2) & (df_c.time_offset_hours==0)]
            f_x = df_x[(df_x.radius==2) & (df_x.time_offset_hours==0)]
            
            # Merge: C + X + T
            merged = pd.merge(f_c[['track_id','mean',target_var]], f_x[['track_id','mean',target_var]], 
                             on=['track_id',target_var], suffixes=('_c', '_x'))
            
            # Filter by Constraint Overlap
            data = merged[(merged['mean_c'] >= overlap[0]) & (merged['mean_c'] <= overlap[1])].copy()
            
            if data.empty:
                ax.text(0.5, 0.5, "No Overlap Data", ha='center', va='center')
                continue

            # --- LOCAL OMEGA BINNING (X-axis) ---
            # Calculate Omega bins based ONLY on the current model's data
            omega_vals = data['mean_x'].values
            x_edges = np.linspace(np.percentile(omega_vals, 1), np.percentile(omega_vals, 99), n_x_bins + 1)
            
            # Binning
            data['x_bin'] = pd.cut(data['mean_x'], bins=x_edges, include_lowest=True)
            data['y_bin'] = pd.cut(data['mean_c'], bins=y_bins, include_lowest=True)
            
            # Aggregate Mean and Count
            pivot_mean = data.groupby(['y_bin', 'x_bin'], observed=False)[target_var].mean().unstack()
            pivot_count = data.groupby(['y_bin', 'x_bin'], observed=False)['track_id'].count().unstack().fillna(0).astype(int)
            
            # --- NORMALIZATION (Per-Model) ---
            data_flat = pivot_mean.values
            if norm_method == 'quantile':
                vmin_model = np.nanquantile(data_flat, p_low)
                vmax_model = np.nanquantile(data_flat, p_high)
            else: # minmax
                vmin_model = np.nanmin(data_flat)
                vmax_model = np.nanmax(data_flat)
            
            denom = vmax_model - vmin_model
            if denom > 0:
                norm_data = (data_flat - vmin_model) / denom
                norm_data = np.clip(norm_data, 0, 1)
            else:
                norm_data = np.zeros_like(data_flat)
                
            # Reconstruct and Sort
            norm_df = pd.DataFrame(norm_data, index=pivot_mean.index, columns=pivot_mean.columns)
            norm_df = norm_df.sort_index(ascending=False)
            pivot_count_sorted = pivot_count.loc[norm_df.index, norm_df.columns]

            # Plot Heatmap
            im = sns.heatmap(norm_df, ax=ax, cmap=cmap_obj, norm=global_norm,
                             cbar=False, annot=pivot_count_sorted, fmt="d", annot_kws={"size": 8})
            
            # Capture the mappable object
            if idx == 0:
                mappable = im.collections[0]
            
            ax.set_title(f"{MODEL_STYLES[model]['label']}\n(Total N={len(data)})")
            ax.set_xlabel(x_var)
            ax.set_ylabel(constrain_var)
            
            # Clean up tick labels
            ax.set_xticklabels([f"{b.mid:.2f}" for b in norm_df.columns], rotation=45)
            y_labels = [f"{b.mid:.1f}" for b in norm_df.index]
            ax.set_yticklabels(y_labels, rotation=0)

        except Exception as e:
            # Note: The merge on 'track_id', 'target_var' might be causing issues if target_var is not identical
            # in both merged dataframes. I simplified it to merge on just 'track_id'.
            print(f"Error plotting {model}: {e}")
            ax.set_visible(False)

    # Turn off unused subplots
    for i in range(len(models), len(axes)):
        axes[i].axis('off')
        
    # Global Colorbar Application
    if mappable:
        cbar_label = f"Normalized {target_var} [0 to 1]"
        fig.colorbar(mappable, ax=axes.tolist(), orientation='vertical', label=cbar_label,
                     shrink=0.8, pad=0.02, aspect=30)
        
    plt.suptitle(f"Normalized Rain vs {x_var} | Constrained {constrain_var}: {overlap[0]:.1f}-{overlap[1]:.1f}mm", y=1.02)
    plt.show()
```

```python
def plot_rain_regimes_hexbin(data_dict, models, crit='all',
                             constrain_var='prw', 
                             x_var='omega850', 
                             target_var='total_total_rain',
                             gridsize=25, # Controls resolution of hexagons
                             cmap='inferno_r'): # 'inferno_r' goes from light (low count) to dark (high count)
    
    # 1. Get Constraints (PRW Overlap)
    overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=constrain_var)
    if not overlap: return
    
    print(f"Constraining {constrain_var} to overlap: {overlap[0]:.2f} - {overlap[1]:.2f}")

    # 2. Determine Global Axis Limits (Extent) for consistent hexbins
    # We scan all models to find the min/max X (Omega) and Y (PRW) to lock the grid
    x_all, y_all = [], []
    
    # Temporary storage to avoid reloading data twice
    loaded_data = {} 

    # Load Dataframes
    for model in models:
        try:
            df_c = data_dict[model][crit][constrain_var]
            df_x = data_dict[model][crit][x_var]
            
            
            # Filter Time/Radius
            f_c = df_c[(df_c.radius==2) & (df_c.time_offset_hours==0)]
            f_x = df_x[(df_x.radius==2) & (df_x.time_offset_hours==0)]
            
            # Merge
            merged = pd.merge(f_c[['track_id','mean',target_var]], f_x[['track_id','mean',target_var]], 
                          on=['track_id',target_var], suffixes=('_c', '_x'))
            # merged = pd.merge(m1, df_t[['track_id', target_var]], on='track_id')
            
            # Filter by Overlap
            data = merged[(merged['mean_c'] >= overlap[0]) & (merged['mean_c'] <= overlap[1])].copy()
            
            if not data.empty:
                loaded_data[model] = data
                x_all.extend(data['mean_x'].values)
                y_all.extend(data['mean_c'].values)
        except: continue
            
    if not loaded_data:
        print("No data found.")
        return

    # Define the Global Extent [xmin, xmax, ymin, ymax]
    # We add a tiny buffer so points on the edge aren't cut off
    extent = [np.min(x_all), np.max(x_all), np.min(y_all), np.max(y_all)]

    # 3. Plotting Setup
    rows = len(models)
    cols = 2
    fig, axes = plt.subplots(rows, cols, figsize=(6, 2*rows),  #figsize=(10, 3.5*rows), 
                             sharex=True, sharey=True, constrained_layout=True)
    
    # If only 1 model, axes is 1D array, make it 2D for consistency
    if rows == 1: axes = np.array([axes])

    for idx, model in enumerate(models):
        if model not in loaded_data:
            axes[idx, 0].text(0.5, 0.5, "No Data", ha='center')
            continue
            
        df = loaded_data[model]
        
        # Calculate Split Threshold (Median P50)
        # You can change this to a fixed value (e.g., 5mm) if you prefer absolute comparisons
        # threshold = df[target_var].median()
        threshold_l = df[target_var].quantile(0.25); threshold_h = df[target_var].quantile(0.75)
        
        # Split Data
        df_low = df[df[target_var] < threshold_l] #df_low = df[df[target_var] <= threshold]
        df_high = df[df[target_var] > threshold_h]
        
        # --- Column 0: Low Intensity ---
        ax_low = axes[idx, 0]
        hb_low = ax_low.hexbin(df_low['mean_x'], df_low['mean_c'], 
                               gridsize=gridsize, extent=extent,
                               cmap=cmap, mincnt=1, linewidths=0.2)
        
        # Annotations
        count_low = len(df_low)
        ax_low.text(0.05, 0.95, f"Low Rain \n(<{threshold_l:.1f}mm)\nN={count_low}", 
                    transform=ax_low.transAxes, va='top', fontsize=9, 
                    bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))
        
        # --- Column 1: High Intensity ---
        ax_high = axes[idx, 1]
        hb_high = ax_high.hexbin(df_high['mean_x'], df_high['mean_c'], 
                                 gridsize=gridsize, extent=extent,
                                 cmap=cmap, mincnt=1, linewidths=0.2)
        
        count_high = len(df_high)
        ax_high.text(0.05, 0.95, f"High Rain \n(>{threshold_h:.1f}mm)\nN={count_high}", 
                     transform=ax_high.transAxes, va='top', fontsize=9,
                     bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))

        # --- Row Labeling ---
        # Add Model Name to the left of the row
        ax_low.set_ylabel(f"{MODEL_STYLES[model]['label']}\n{constrain_var}", fontsize=11, fontweight='bold')
        
        # Add colorbars for density (optional, but helpful per subplot)
        # Using inset_axes to place it inside or tiny next to plot
        # For cleanliness, we rely on the density color relative to itself
        
    # 4. Final Formatting
    # Set X-labels only on bottom row
    for ax in axes[-1, :]:
        ax.set_xlabel(x_var)

    # Set Column Titles
    axes[0, 0].set_title(f"Less Intense Precip (< P25)", fontsize=10.5)
    axes[0, 1].set_title(f"Most Intense Precip (> P75)", fontsize=10.5)
    
    plt.suptitle(f"Phase Space Density: {constrain_var} vs {x_var}(Constrained {constrain_var}: {overlap[0]:.1f}-{overlap[1]:.1f})", y=1.03, fontsize=10)
    plt.show()

# Usage:
# plot_rain_regimes_hexbin(processed_data, ['scream', 'um', 'icon'], crit='min2_longremoved')
```

```python
def plot_normalized_rain_omega_hexbin(data_dict, models, crit='all',
                                      constrain_var='prw', manual_prw_limits=(52, 53),
                                      x_var='omega850', 
                                      y_var='total_total_rain',
                                      gridsize=20, 
                                      cmap='Blues'):
    
    print(f"Applying Manual Constraint: {constrain_var} between {manual_prw_limits[0]} and {manual_prw_limits[1]}")

    # Setup Plot Grid
    n_models = len(models)
    cols = 3
    rows = (n_models // cols) + (1 if n_models % cols > 0 else 0)
    
    fig, axes = plt.subplots(rows, cols, figsize=(3*cols, 3*rows), constrained_layout=True)
    axes = axes.flatten()

    for idx, model in enumerate(models):
        ax = axes[idx]
        
        try:
            # 1. Load FULL Data (to calculate Global Min/Max for Normalization)
            # We need the full distribution to know what "0" and "1" mean for this model
            df_c_full = data_dict[model][crit][constrain_var]
            df_x_full = data_dict[model][crit][x_var]
            # df_y_full = data_dict[model][crit][y_var] # Rain
            
            if df_x_full is None: continue

            # Calculate Global Min/Max for Normalization (Per Model)
            # Using 1st and 99th percentile is often safer than absolute min/max to avoid outliers,
            # but here I use absolute min/max as requested.
            x_min, x_max = df_x_full['mean'].quantile(0.05), df_x_full['mean'].quantile(0.95)
            y_min, y_max = df_c_full[y_var].quantile(0.05) , df_c_full[y_var].quantile(0.95) # 
            
            # 2. Filter Data based on Manual Constraint
            # Filter Time/Radius first
            f_c = df_c_full[(df_c_full.radius==2) & (df_c_full.time_offset_hours==0)]
            f_x = df_x_full[(df_x_full.radius==2) & (df_x_full.time_offset_hours==0)]
            
            # Merge
            merged = pd.merge(f_c[['track_id','mean',y_var]], f_x[['track_id','mean',y_var]], 
                          on=['track_id',y_var], suffixes=('_c', '_x'))
            # merged = pd.merge(m1, df_y_full[['track_id', y_var]], on='track_id')
            
            # Apply Manual PRW Constraint
            subset = merged[
                (merged['mean_c'] >= manual_prw_limits[0]) & 
                (merged['mean_c'] <= manual_prw_limits[1])
            ].copy()
            
            if subset.empty:
                ax.text(0.5, 0.5, "No Data in Range", ha='center', va='center')
                continue
                
            # 3. Normalize the SUBSET using the GLOBAL limits
            # This places the specific PRW slice into the context of the whole model
            subset['norm_x'] = (subset['mean_x'] - x_min) / (x_max - x_min)
            subset['norm_y'] = (subset[y_var] - y_min) / (y_max - y_min)
            
            # Clip to 0-1 just in case
            subset['norm_x'] = subset['norm_x'].clip(0, 1)
            subset['norm_y'] = subset['norm_y'].clip(0, 1)

            # 4. Plot Hexbin
            # X = Normalized Omega, Y = Normalized Rain, C = Density (default)
            hb = ax.hexbin(subset['norm_x'], subset['norm_y'], 
                           gridsize=gridsize, 
                           cmap=cmap, 
                           extent=[0, 1, 0, 1], # Force plotting area to normalized unit square
                           mincnt=1, 
                           linewidths=0.2)
            
            # 5. Annotations and Style
            ax.set_title(f"{MODEL_STYLES[model]['label']}\n(N={len(subset)})")
            
            # Add physical values labels to corners so we know what 0 and 1 mean
            ax.text(0.02, -0.02, f"X_min:\n{x_min:.2f}", transform=ax.transAxes, 
                    fontsize=7, va='top', ha='left', color='gray')
            ax.text(0.98, -0.02, f"X_max:\n{x_max:.2f}", transform=ax.transAxes, 
                    fontsize=7, va='top', ha='right', color='gray')
            ax.text(-0.02, 0.98, f"Y_max:\n{y_max:.1f}", transform=ax.transAxes, 
                    fontsize=7, va='top', ha='right', rotation=90, color='gray')

            ax.set_xlim(0, 1)
            ax.set_ylim(0, 1)
            
            # Only put axis labels on the edges
            if idx >= len(models) - cols:
                ax.set_xlabel(f"Norm. {x_var}")
            if idx % cols == 0:
                ax.set_ylabel(f"Norm. {y_var}")

        except Exception as e:
            print(f"Error {model}: {e}")

    # Hide unused axes
    for i in range(len(models), len(axes)):
        axes[i].axis('off')

    plt.suptitle(f"Norm. Rain vs Omega | Fixed {constrain_var}: {manual_prw_limits[0]}-{manual_prw_limits[1]}", y=1.02)
    plt.show()

# --- Usage Example ---
# plot_normalized_rain_omega_hexbin(processed_data, 
#                                   ['scream', 'um', 'icon', 'ifs', 'nicam', 'obs'], 
#                                   manual_prw_limits=(52, 53),
#                                   crit='min2_longremoved')
```

```python
def plot_mean_rain_vs_binned_omega(data_dict, models, crit='all',
                                   constrain_var='prw', 
                                   manual_prw_limits=None,  # None = use auto-calculated overlap
                                   omega_var='omega850', 
                                   rain_var='total_total_rain',
                                   stat='mean', 
                                   radius=2, 
                                   time=0,
                                   omega_qtiles = [0, 0.25, 0.5, 0.75, 1.0],
                                   normalize=False,  # Normalize rain and omega per model
                                   use_quantile_norm=False,  # Normalize rain and omega per model
                                   lower_quantile=0.05,      
                                   upper_quantile=0.95,  clip_bins_to_quantiles=False,
                                   min_samples_per_bin=10,  # Minimum samples required per bin
                                   legend_fontsize=10, 
                                   ymin=None, xmin=None,
                                   ymax=None, xmax=None,
                                   figsize=(10, 8)):
    """
    Plot mean/median rain vs binned omega850, constrained by PRW range.
    
    Parameters:
    -----------
    data_dict : dict
        Nested dictionary with structure: data_dict[model][crit][var]
    models : list
        List of model names to include in the plot
    crit : str
        Criteria key (e.g., 'all', 'min2_longremoved')
    constrain_var : str
        Variable to constrain (e.g., 'prw')
    manual_prw_limits : tuple or None
        If tuple (min, max), use these limits. If None, auto-calculate using get_common_overlap
    omega_var : str
        Variable to bin on x-axis (e.g., 'omega850')
    rain_var : str
        Target variable to plot (e.g., 'total_total_rain')
    stat : str
        Statistic to compute ('mean' or 'median')
    radius : int
        Radius filter value
    time : int
        Time offset hours filter value
    normalize : bool
        If True, normalize rain (0=min, 1=max) and omega (1=strongest ascent, 0=weakest) per model
    min_samples_per_bin : int
        Minimum number of samples required per bin to display
    legend_fontsize : int
        Font size for legend
    ymin, ymax : float or None
        Y-axis limits for main plot and rain distribution
    figsize : tuple
        Figure size (width, height)
    """
    
    # 1. Determine PRW constraint range
    if manual_prw_limits is None:
        # Auto-calculate using get_common_overlap
        overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=constrain_var)
        if not overlap:
            print("Could not determine common PRW overlap.")
            return
        prw_limits = overlap
        print(f"Using auto-calculated {constrain_var} range: {prw_limits[0]:.2f} - {prw_limits[1]:.2f}")
    else:
        prw_limits = manual_prw_limits
        print(f"Using manual {constrain_var} range: {prw_limits[0]:.2f} - {prw_limits[1]:.2f}")
    
    # 2. Collect all omega values within PRW constraint for quartile calculation
    all_omega_vals = []
    all_data_for_models = {}  # Store for later processing

    for model in models:
        try:
            df_c = data_dict[model][crit][constrain_var]
            df_x = data_dict[model][crit][omega_var]
            
            if df_c is None or df_x is None:
                continue
            
            # Filter Time/Radius
            f_c = df_c[(df_c.radius == radius) & (df_c.time_offset_hours == time)]
            f_x = df_x[(df_x.radius == radius) & (df_x.time_offset_hours == time)]
            
            # Merge on track_id and rain_var
            merged = pd.merge(
                f_c[['track_id', 'mean', rain_var]], 
                f_x[['track_id', 'mean']], 
                on='track_id', 
                suffixes=('_c', '_x')
            )
            
            # Apply PRW Constraint
            subset = merged[
                (merged['mean_c'] >= prw_limits[0]) & 
                (merged['mean_c'] <= prw_limits[1])
            ].copy()
            
            if not subset.empty:
                omega_vals = subset['mean_x'].values

                if clip_bins_to_quantiles:
                    q_low = np.quantile(omega_vals, lower_quantile)
                    q_high = np.quantile(omega_vals, upper_quantile)
                    omega_vals_for_bins = np.clip(omega_vals, q_low, q_high)
                else:
                    omega_vals_for_bins = omega_vals
                
                all_omega_vals.extend(omega_vals_for_bins)
                all_data_for_models[model] = subset
                
        except Exception as e:
            print(f"Warning: Error loading data for {model}: {e}")
            continue

    if not all_omega_vals:
        print("No data found in the specified PRW range.")
        return
    
    # 3. Calculate quartile-based omega bins (4 bins) BEFORE normalization
    quartiles = omega_qtiles
    omega_edges_raw = np.quantile(all_omega_vals, quartiles)
    
    # Ensure unique edges (in case of duplicate values)
    omega_edges_raw = np.unique(omega_edges_raw)
    
    if len(omega_edges_raw) < 2:
        print("Not enough unique omega values to create bins.")
        return
    
    print(f"Omega bins (raw quartiles): {omega_edges_raw}")
    
    # 4. Calculate normalization ranges per model if needed
    model_norm_ranges = {}
    if normalize:
        print("\n--- Normalization Ranges ---")
        for model in models:
            if model not in all_data_for_models:
                continue
            subset = all_data_for_models[model]

            if use_quantile_norm:
                # For rain: min to max (0 = min, 1 = max)
                rain_min = subset[rain_var].quantile(0.05) #min()
                rain_max = subset[rain_var].quantile(0.95) #.max()
                
                # For omega: reverse since negative = strong ascent
                # 1 = strongest ascent (most negative), 0 = weakest (least negative/most positive)
                omega_min = subset['mean_x'].quantile(0.05) #min()  # least negative (or most positive)
                omega_max = subset['mean_x'].quantile(0.95) #max()  # most positive
            else:
                # For rain: min to max (0 = min, 1 = max)
                rain_min = subset[rain_var].min()
                rain_max = subset[rain_var].max()

                # For omega: reverse since negative = strong ascent
                # 1 = strongest ascent (most negative), 0 = weakest (least negative/most positive)
                omega_min = subset['mean_x'].min()  # least negative (or most positive)
                omega_max = subset['mean_x'].max() #max()  # most positive
            
            model_norm_ranges[model] = {
                'rain_min': rain_min,
                'rain_max': rain_max,
                'omega_min': omega_min,
                'omega_max': omega_max
            }
            print(f"  {MODEL_STYLES[model]['label']}: Rain [{rain_min:.2f}, {rain_max:.2f}], Omega [{omega_min:.3f}, {omega_max:.3f}]")
    
    # 5. Process data per model and calculate statistics per bin
    processed_dfs = []
    all_rain_data = []  # For overall distribution plot
    all_omega_data = []  # For omega distribution plot
    model_sample_counts = {}  # Track total samples per model

    for model in models:
        if model not in all_data_for_models:
            continue
            
        subset = all_data_for_models[model].copy()
        model_sample_counts[model] = len(subset)
        
        # Normalize if requested
        if normalize:
            ranges = model_norm_ranges[model]
            
            if use_quantile_norm:
                # Use quantiles (robust to outliers)
                rain_min_norm = ranges['rain_min']
                rain_max_norm = ranges['rain_max']
                clipped_rain = np.clip(subset[rain_var], rain_min_norm, rain_max_norm)
                subset[rain_var] = ((clipped_rain - rain_min_norm) / (rain_max_norm - rain_min_norm))
                # Normalize omega (reverse: 1 = strongest ascent/most negative, 0 = weakest)
                omega_min_norm = ranges['omega_min']
                omega_max_norm = ranges['omega_max']
                clipped_omega = np.clip(subset['mean_x'], omega_min_norm, omega_max_norm)
                subset['mean_x'] = 1 - ((clipped_omega - omega_min_norm) / (omega_max_norm - omega_min_norm))
                
            else:
                # Normalize rain (0 to 1)
                if ranges['rain_max'] > ranges['rain_min']:
                    subset[rain_var] = (subset[rain_var] - ranges['rain_min']) / (ranges['rain_max'] - ranges['rain_min'])
                # Use min/max (original behavior)
                if ranges['omega_max'] > ranges['omega_min']:
                    subset['mean_x'] = 1 - ((subset['mean_x'] - ranges['omega_min']) / (ranges['omega_max'] - ranges['omega_min']))
            
            # For normalized data, use fixed bins from 0 to 1
            omega_edges = np.linspace(0, 1, 5)  # 4 equal bins in normalized space
        else:
            # Use the raw quartile bins
            omega_edges = omega_edges_raw
        
        # Store all rain data for distribution plot
        rain_data_model = subset[[rain_var]].copy()
        rain_data_model['Model'] = model
        rain_data_model['Model_Label'] = MODEL_STYLES[model]['label']
        all_rain_data.append(rain_data_model)
        
        # Store all omega data for distribution plot
        omega_data_model = subset[['mean_x']].copy()
        omega_data_model['Model'] = model
        omega_data_model['Model_Label'] = MODEL_STYLES[model]['label']
        all_omega_data.append(omega_data_model)
        
        # Create bins
        omega_bins = pd.cut(subset['mean_x'], bins=omega_edges, include_lowest=True, duplicates='drop')
        
        # Group by bin and calculate statistics
        agg_results = subset.groupby(omega_bins, observed=False)[rain_var].agg([stat, 'count'])
        
        # Filter bins with insufficient samples
        agg_results = agg_results[agg_results['count'] >= min_samples_per_bin].copy()
        
        if agg_results.empty:
            print(f"Warning: {model} has no bins with >= {min_samples_per_bin} samples")
            continue
        
        # Calculate bin centers for plotting - ENSURE FLOAT TYPE
        agg_results['omega_center'] = agg_results.index.map(lambda x: x.mid if pd.notna(x) else np.nan).astype(float)
        
        # Add metadata
        agg_results['Model'] = model
        agg_results['Model_Label'] = MODEL_STYLES[model]['label']
        
        # Reset index to make concatenation easier
        agg_results = agg_results.reset_index(drop=True)
        
        processed_dfs.append(agg_results)

    if not processed_dfs:
        print("No models had sufficient data in the constrained range.")
        return
    
    final_df = pd.concat(processed_dfs, ignore_index=True)
    
    # Combine all rain data for distribution
    all_rain_combined = pd.concat(all_rain_data, ignore_index=True)
    
    # Combine all omega data for distribution
    all_omega_combined = pd.concat(all_omega_data, ignore_index=True)
    
    # 6. Create figure with joint plot layout (main plot + marginal distributions)
    # Using gridspec for custom layout: omega dist on top, main plot in center, rain dist on right
    fig = plt.figure(figsize=figsize)
    gs = fig.add_gridspec(2, 2, width_ratios=[4, 1], height_ratios=[1, 4],
                          hspace=0.1, wspace=0.1)
    
    ax_main = fig.add_subplot(gs[1, 0])  # Main scatter/line plot
    ax_top = fig.add_subplot(gs[0, 0], sharex=ax_main)  # Omega distribution (top)
    ax_right = fig.add_subplot(gs[1, 1], sharey=ax_main)  # Rain distribution (right)
    
    # 7. Plot TOP: Omega distribution (KDE)
    print(f"\n=== Plotting omega distribution (top panel) ===")
    print(f"Total omega data points: {len(all_omega_combined)}")
    for model in models:
        if model not in all_data_for_models:
            continue
        model_omega = all_omega_combined[all_omega_combined['Model'] == model]
        
        if len(model_omega) > 0:
            # print(f"  {MODEL_STYLES[model]['label']}: {len(model_omega)} omega values, range [{model_omega['mean_x'].min():.3f}, {model_omega['mean_x'].max():.3f}]")
            if normalize:
                sns.kdeplot(data=model_omega,x='mean_x',color=MODEL_STYLES[model]['color'],
                linewidth=MODEL_STYLES[model].get('lw', 2.5),clip=(0,1),
                ax=ax_top,label=MODEL_STYLES[model]['label'],fill=False)
            else:
                sns.kdeplot(data=model_omega,x='mean_x',color=MODEL_STYLES[model]['color'],
                linewidth=MODEL_STYLES[model].get('lw', 2.5), ax=ax_top,label=MODEL_STYLES[model]['label'],fill=False)
    
    ax_top.set_xlabel('')  # Remove x-label (shared with main plot)
    ax_top.set_ylabel('Density', fontsize=10)
    ax_top.tick_params(axis='x', labelbottom=False)  # Hide x-axis tick labels
    ax_top.grid(True, linestyle=':', alpha=0.6, axis='x')
    ax_top.legend().set_visible(False)  # Hide legend (already on main plot)
    
    if normalize:
        ax_top.set_title('Omega Distribution (Normalized)', fontsize=11)
        ax_top.set_xlim(0, 1.05)
    else:
        ax_top.set_title('Omega Distribution', fontsize=11)
    
    # 8. Plot MAIN: Mean/Median rain vs omega bins
    for model in models:
        if model not in all_data_for_models:
            continue
        
        model_data = final_df[final_df['Model'] == model]
        if model_data.empty:
            continue
            
        model_label = MODEL_STYLES[model]['label']
        n_samples = model_sample_counts[model]
        
        # Add sample count to label
        label_with_count = f"{model_label} (N={n_samples})"
        
        ax_main.plot(
            model_data['omega_center'], 
            model_data[stat],
            color=MODEL_STYLES[model]['color'],
            linewidth=MODEL_STYLES[model].get('lw', 2.5),
            label=label_with_count,
            marker='o',
            markersize=6
        )
    
    # Add vertical line at omega=0 (only if not normalized)
    if not normalize:
        ax_main.axvline(0, color='gray', linestyle='--', alpha=0.7, linewidth=1.5)
    
    # Set labels and title based on normalization
    if normalize:
        ax_main.set_xlabel(f"Normalized {omega_var}\n(1=strongest ascent, 0=weakest)", fontsize=11)
        ax_main.set_ylabel(f"Normalized {stat.capitalize()} {rain_var}\n(0=min, 1=max)", fontsize=11)
        ax_main.set_ylim(0, 1.05)
        ax_main.set_xlim(0, 1.05)
    else:
        ax_main.set_xlabel(f"{omega_var} (Pa/s)", fontsize=12)
        ax_main.set_ylabel(f"{stat.capitalize()} {rain_var} (mm)", fontsize=12)
    
    # Apply custom Y limits if provided
    if ymin is not None or ymax is not None:
        current_ylim = ax_main.get_ylim()
        new_ymin = ymin if ymin is not None else current_ylim[0]
        new_ymax = ymax if ymax is not None else current_ylim[1]
        ax_main.set_ylim(new_ymin, new_ymax)

    if xmin is not None or xmax is not None:
        current_xlim = ax_main.get_xlim()
        new_xmin = xmin if xmin is not None else current_xlim[0]
        new_xmax = xmax if xmax is not None else current_xlim[1]
        ax_main.set_xlim(new_xmin, new_xmax)
    
    ax_main.legend(fontsize=legend_fontsize, frameon=True, loc='best')
    ax_main.grid(True, linestyle=':', alpha=0.6)
    
    # 9. Plot RIGHT: Overall rain distribution (KDE)
    for model in models:
        if model not in all_data_for_models:
            continue
        model_rain = all_rain_combined[all_rain_combined['Model'] == model]
        
        if len(model_rain) > 0:
            sns.kdeplot(
                data=model_rain,
                y=rain_var,
                color=MODEL_STYLES[model]['color'],
                linewidth=MODEL_STYLES[model].get('lw', 2.5),
                ax=ax_right,
                label=MODEL_STYLES[model]['label']
            )
    
    ax_right.set_ylabel('')  # Remove y-label (shared with main plot)
    ax_right.set_xlabel('Density', fontsize=10)
    ax_right.tick_params(axis='y', labelleft=False)  # Hide y-axis tick labels
    ax_right.grid(True, linestyle=':', alpha=0.6, axis='y')
    ax_right.legend().set_visible(False)  # Hide legend (already on main plot)
    
    if normalize:
        ax_right.set_title('Rain Dist.\n(Normalized)', fontsize=11)
    else:
        ax_right.set_title('Rain Distribution', fontsize=11)

    # ========== NEW: Add regression statistics from binned data ==========
    from scipy import stats
    
    regression_stats = {}
    
    # Calculate regression from the plotted binned points
    for model in models:
        if model not in model_sample_counts:
            continue
        
        # Get the binned data for this model from final_df
        model_binned = final_df[final_df['Model'] == model].copy()
        
        if len(model_binned) < 2:
            continue
        
        # Extract omega bin centers and mean/median rain values
        omega_centers = model_binned['omega_center'].values
        rain_values = model_binned[stat].values
        
        # Remove any NaN values
        valid_mask = ~(np.isnan(omega_centers) | np.isnan(rain_values))
        omega_clean = omega_centers[valid_mask]
        rain_clean = rain_values[valid_mask]
        
        if len(omega_clean) >= 2:
            slope, intercept, r_value, p_value, std_err = stats.linregress(omega_clean, rain_clean)
            regression_stats[model] = {
                'slope': slope,
                'intercept': intercept,
                'r2': r_value ** 2,
                'pval': p_value,
                'n_bins': len(omega_clean)
            }
    
    # Create text annotation box with regression statistics
    if regression_stats:
        if normalize:
            reg_text = "Linear fit (binned data):\n"
            units = ""
        else:
            reg_text = f"Linear fit (binned data):\n"
            units = " mm·s/Pa"
        
        for model in models:
            if model in regression_stats:
                stats_dict = regression_stats[model]
                if stats_dict['pval'] < 0.001:
                    p_str = "p<0.001"
                elif stats_dict['pval'] < 0.01:
                    p_str = "p<0.01"
                else:
                    p_str = f"p={stats_dict['pval']:.3f}"
                
                reg_text += (f"{MODEL_STYLES[model]['label']}: "
                           f"slope={stats_dict['slope']:.3f}{units}, "
                           f"R²={stats_dict['r2']:.2f}, {p_str}\n")
        
        # Add text box to main plot
        ax_main.text(0.02, 0.98, reg_text.strip(),
                    transform=ax_main.transAxes,
                    fontsize=legend_fontsize - 1,
                    verticalalignment='top',
                    bbox=dict(boxstyle='round', facecolor='white', alpha=0.8, edgecolor='gray'))
    
    # ========== END NEW CODE ==========

    # Add overall title
    if normalize:
        fig.suptitle(f"Normalized {stat.capitalize()} Rain vs {omega_var} ({constrain_var}: {prw_limits[0]:.1f}-{prw_limits[1]:.1f} mm)", 
                     fontsize=12, y=0.98)
    else:
        fig.suptitle(f"{stat.capitalize()} Rain vs {omega_var} ({constrain_var}: {prw_limits[0]:.1f}-{prw_limits[1]:.1f} mm)", 
                     fontsize=12, y=0.98)
    
    plt.tight_layout()
    plt.show()
    
    return fig, final_df, regression_stats
```

```python
# Investigate why obs and obs_v7 have different normalized omega distributions
# even though their raw distributions are similar
# Setup
models_to_compare = ['obs', 'obs_v7','scream']
crit = 'min2_longremoved'
constrain_var = 'prw'
omega_var = 'omega850'
prw_limits = (50, 60)  # Your specified range
radius = 2
time = 0

print("="*70)
print("INVESTIGATING NORMALIZATION DIFFERENCE: obs vs obs_v7")
print("="*70)

# Extract data for both models
model_data = {}

for model in models_to_compare:
    df_c = processed_data[model][crit][constrain_var]
    df_x = processed_data[model][crit][omega_var]
    
    # Filter Time/Radius
    f_c = df_c[(df_c.radius == radius) & (df_c.time_offset_hours == time)]
    f_x = df_x[(df_x.radius == radius) & (df_x.time_offset_hours == time)]
    
    # Merge
    merged = pd.merge(
        f_c[['track_id', 'mean']], 
        f_x[['track_id', 'mean']], 
        on='track_id', 
        suffixes=('_prw', '_omega')
    )
    
    # Apply PRW constraint
    subset = merged[
        (merged['mean_prw'] >= prw_limits[0]) & 
        (merged['mean_prw'] <= prw_limits[1])
    ].copy()
    
    model_data[model] = subset
    
    # Print statistics
    print(f"\n{MODEL_STYLES[model]['label']}:")
    print(f"  Number of samples in PRW range ({prw_limits[0]}-{prw_limits[1]} mm): {len(subset)}")
    print(f"  Raw omega850 range: [{subset['mean_omega'].min():.4f}, {subset['mean_omega'].max():.4f}]")
    print(f"  Raw omega850 mean: {subset['mean_omega'].mean():.4f}")
    print(f"  Raw omega850 std: {subset['mean_omega'].std():.4f}")
    print(f"  Raw omega850 median: {subset['mean_omega'].median():.4f}")
    print(f"  Raw omega850 quartiles: 25%={subset['mean_omega'].quantile(0.25):.4f}, "
          f"50%={subset['mean_omega'].quantile(0.5):.4f}, 75%={subset['mean_omega'].quantile(0.75):.4f}")

# Compare the distributions
print("\n" + "="*70)
print("KEY INSIGHT:")
print("="*70)

# obs_range = model_data['obs']['mean_omega'].max() - model_data['obs']['mean_omega'].min()
# obs_v7_range = model_data['obs_v7']['mean_omega'].max() - model_data['obs_v7']['mean_omega'].min()
obs_range = model_data['obs']['mean_omega'].quantile(0.05) - model_data['obs']['mean_omega'].quantile(0.05)
obs_v7_range = model_data['obs_v7']['mean_omega'].quantile(0.95) - model_data['obs_v7']['mean_omega'].quantile(0.05)

print(f"\nobs omega850 range width: {obs_range:.4f}")
print(f"obs_v7 omega850 range width: {obs_v7_range:.4f}")
print(f"Range ratio (obs/obs_v7): {obs_range/obs_v7_range:.2f}")

print("\nNORMALIZATION FORMULA: normalized = 1 - (value - min) / (max - min)")
print("  → This maps strongest ascent (most negative) to 1")
print("  → And weakest ascent (least negative/most positive) to 0")

# Apply normalization to see the effect
for model in models_to_compare:
    omega_vals = model_data[model]['mean_omega'].values
    omega_min = np.quantile(omega_vals,0.05) #min()
    omega_max = np.quantile(omega_vals,0.95) #max()
    
    clipped = np.clip(omega_vals, omega_min, omega_max)
    # Normalize (reversed: 1 = strongest ascent)
    normalized = 1 - ((clipped - omega_min) / (omega_max - omega_min))
    
    model_data[model]['normalized_omega'] = normalized
    
    print(f"\n{MODEL_STYLES[model]['label']} after normalization:")
    print(f"  Normalized range: [{normalized.min():.4f}, {normalized.max():.4f}]")
    print(f"  Normalized mean: {normalized.mean():.4f}")
    print(f"  Normalized std: {normalized.std():.4f}")
    print(f"  Normalized median: {np.median(normalized):.4f}")

# Create comparison plots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: Raw omega distributions
ax = axes[0, 0]
for model in models_to_compare:
    sns.kdeplot(
        data=model_data[model],
        x='mean_omega',
        ax=ax,
        color=MODEL_STYLES[model]['color'],
        linewidth=2.5,
        label=MODEL_STYLES[model]['label'],
        fill=True,
        alpha=0.3
    )
ax.set_xlabel('Omega850 (Pa/s)', fontsize=12)
ax.set_ylabel('Density', fontsize=12)
ax.set_title('Raw Omega850 Distributions\n(PRW constrained to 50-60 mm)', fontsize=13, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
ax.axvline(0, color='black', linestyle='--', alpha=0.5, label='Zero line')

# Plot 2: Raw omega histograms
ax = axes[0, 1]
for model in models_to_compare:
    ax.hist(
        model_data[model]['mean_omega'],
        bins=30,
        alpha=0.5,
        color=MODEL_STYLES[model]['color'],
        label=MODEL_STYLES[model]['label'],
        density=True
    )
ax.set_xlabel('Omega850 (Pa/s)', fontsize=12)
ax.set_ylabel('Density', fontsize=12)
ax.set_title('Raw Omega850 Histograms', fontsize=13, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

# Plot 3: Normalized omega distributions
ax = axes[1, 0]
for model in models_to_compare:
    sns.kdeplot(
        data=model_data[model],
        x='normalized_omega',
        ax=ax,
        color=MODEL_STYLES[model]['color'],
        linewidth=2.5,
        label=MODEL_STYLES[model]['label'],
        fill=False,
        clip=(0, 1)
    )
ax.set_xlabel('Normalized Omega850\n(1=strongest ascent, 0=weakest)', fontsize=12)
ax.set_ylabel('Density', fontsize=12)
ax.set_title('Normalized Omega850 Distributions\n(Different ranges cause different shapes!)', fontsize=13, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)

# Plot 4: Box plots comparison
ax = axes[1, 1]
combined_raw = pd.concat([
    model_data['obs'][['mean_omega']].assign(Model=MODEL_STYLES['obs']['label'], Type='Raw'),
    model_data['obs_v7'][['mean_omega']].assign(Model=MODEL_STYLES['obs_v7']['label'], Type='Raw')
])
combined_norm = pd.concat([
    model_data['obs'][['normalized_omega']].rename(columns={'normalized_omega': 'mean_omega'}).assign(Model=MODEL_STYLES['obs']['label'], Type='Normalized'),
    model_data['obs_v7'][['normalized_omega']].rename(columns={'normalized_omega': 'mean_omega'}).assign(Model=MODEL_STYLES['obs_v7']['label'], Type='Normalized')
])
combined = pd.concat([combined_raw, combined_norm])

sns.boxplot(data=combined, x='Type', y='mean_omega', hue='Model', ax=ax, palette=[MODEL_STYLES['obs']['color'], MODEL_STYLES['obs_v7']['color']])
ax.set_ylabel('Value', fontsize=12)
ax.set_xlabel('', fontsize=12)
ax.set_title('Box Plot Comparison', fontsize=13, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# # Print explanation
# print("\n" + "="*70)
# print("EXPLANATION:")
# print("="*70)
# print("""
# When normalization is applied PER MODEL:
# 1. Each model's omega values are normalized to [0, 1] based on ITS OWN min/max
# 2. If obs and obs_v7 have different ranges (even slightly), the normalization 
#    stretches/compresses the distributions differently
# 3. Example:
#    - If obs has range [-0.5, 0.1] (width=0.6)
#    - And obs_v7 has range [-0.4, 0.1] (width=0.5)
#    - Then the same raw value (e.g., -0.2) will map to different normalized values
   
# This is why SIMILAR raw distributions become DIFFERENT after normalization!

# The raw distributions look similar because they have similar shapes and central 
# tendencies, but their min/max extremes differ, causing different normalization 
# transformations.
# """)
```

```python
# Check if omega extremes coincide with rain extremes
models_to_check = ['obs', 'obs_v7', 'scream']
crit = 'min2_longremoved'
prw_limits = (50, 60)
extreme_quantile = 0.05

print("="*80)
print("DO OMEGA EXTREMES COINCIDE WITH RAIN EXTREMES?")
print("="*80)

for model in models_to_check:
    # Get data
    df_c = processed_data[model][crit]['prw']
    df_x = processed_data[model][crit]['omega850']
    f_c = df_c[(df_c.radius == 2) & (df_c.time_offset_hours == 0)]
    f_x = df_x[(df_x.radius == 2) & (df_x.time_offset_hours == 0)]
    merged = pd.merge(f_c[['track_id', 'mean', 'total_total_rain']], 
                     f_x[['track_id', 'mean']], 
                     on='track_id', suffixes=('_prw', '_omega'))
    subset = merged[(merged['mean_prw'] >= prw_limits[0]) & 
                    (merged['mean_prw'] <= prw_limits[1])].copy()
    
    # Identify extremes
    omega_low = subset['mean_omega'].quantile(extreme_quantile)
    omega_high = subset['mean_omega'].quantile(1 - extreme_quantile)
    rain_low = subset['total_total_rain'].quantile(extreme_quantile)
    rain_high = subset['total_total_rain'].quantile(1 - extreme_quantile)
    
    # Flag extreme cases
    omega_extreme = (subset['mean_omega'] < omega_low) | (subset['mean_omega'] > omega_high)
    rain_extreme = (subset['total_total_rain'] < rain_low) | (subset['total_total_rain'] > rain_high)
    
    # Calculate overlap
    both_extreme = omega_extreme & rain_extreme
    only_omega = omega_extreme & ~rain_extreme
    only_rain = rain_extreme & ~omega_extreme
    neither = ~omega_extreme & ~rain_extreme
    
    print(f"\n{MODEL_STYLES[model]['label']} (N={len(subset)}):")
    print(f"  Omega extremes: {omega_extreme.sum()} ({omega_extreme.sum()/len(subset)*100:.1f}%)")
    print(f"  Rain extremes:  {rain_extreme.sum()} ({rain_extreme.sum()/len(subset)*100:.1f}%)")
    print(f"  Both extreme:   {both_extreme.sum()} ({both_extreme.sum()/len(subset)*100:.1f}%)")
    print(f"  Only omega:     {only_omega.sum()} ({only_omega.sum()/len(subset)*100:.1f}%)")
    print(f"  Only rain:      {only_rain.sum()} ({only_rain.sum()/len(subset)*100:.1f}%)")
    print(f"  Neither:        {neither.sum()} ({neither.sum()/len(subset)*100:.1f}%)")
    
    # Correlation between omega and rain
    corr = subset['mean_omega'].corr(subset['total_total_rain'])
    print(f"  Omega-Rain correlation: {corr:.3f}")
    
    # Check if extreme omega → extreme rain
    if omega_extreme.sum() > 0:
        mean_rain_omega_extreme = subset[omega_extreme]['total_total_rain'].mean()
        mean_rain_normal = subset[~omega_extreme]['total_total_rain'].mean()
        print(f"  Mean rain (omega extreme): {mean_rain_omega_extreme:.1f} mm")
        print(f"  Mean rain (omega normal):  {mean_rain_normal:.1f} mm")
```

```python
def plot_rain_vs_omega_constrained(data_dict, models, crit='all', 
                                   constrain_var='prw', 
                                   x_var='omega850', 
                                   y_var='total_total_rain',  radius=2, time=0):
    
    # 1. Get constraint range (Intersection of 10th-90th percentiles)
    overlap, _ = get_common_overlap(data_dict, models, crit=crit, var=constrain_var)
    if not overlap: return

    plot_data = []

    for model in models:
        # We need to merge the Constrain Var DF and the X Var DF
        # Assuming they share 'track_id', 'radius', 'time_offset_hours'
        
        df_constrain = data_dict[model][crit][constrain_var]
        df_x = data_dict[model][crit][x_var]
        
        # Merge on index or common columns. 
        # Crucial: Ensure 'radius' and 'time' are filtered BEFORE merge to avoid duplicates
        # or merge on them too.
        
        # Simple approach: Filter both first
        f_c = df_constrain[(df_constrain.radius==radius) & (df_constrain.time_offset_hours==time)]
        f_x = df_x[(df_x.radius==radius) & (df_x.time_offset_hours==time)]
        
        # Merge. We need 'track_id' to match the exact same storm event
        merged = pd.merge(f_c[['track_id', 'mean', y_var]], 
                          f_x[['track_id', 'mean']], 
                          on='track_id', suffixes=('_constrain', '_x'))
        
        # Filter by Constraint (e.g., PRW between 50 and 55)
        filtered = merged[
            (merged['mean_constrain'] >= overlap[0]) & 
            (merged['mean_constrain'] <= overlap[1])
        ].copy()
        
        filtered['Model'] = MODEL_STYLES[model]['label']
        plot_data.append(filtered)

    full_df = pd.concat(plot_data)

    # 2. Bin the X-variable (Omega)
    # create 10 bins for Omega
    full_df['omega_bin'] = pd.cut(full_df['mean_x'], bins=10)

    # 3. Plot
    plt.figure(figsize=(12, 6))
    
    # Boxplot: X=Omega Bin, Y=Rain, Hue=Model
    sns.boxplot(data=full_df, x='omega_bin', y=y_var, hue='Model', 
                palette={v['label']: v['color'] for k,v in MODEL_STYLES.items() if k in models},
                showfliers=False) # Hide outliers to see the distribution box better
    
    plt.title(f"Rain vs Omega850 (Constrained: {overlap[0]:.1f} < {constrain_var} < {overlap[1]:.1f})")
    plt.xticks(rotation=45)
    plt.xlabel("Omega 850hPa Bins (Pa/s)")
    plt.ylabel("Total Rainfall Distribution")
    plt.show()
```

### Correlation Matrix Analysis

Analyze correlations between environmental variables within a specific time window

#### Function: `plot_correlation_matrix`

This function computes and visualizes correlations between environmental variables within a specified time window.

**Key Features:**
- Filter data by time window (e.g., pre-convective: -6 to 0 hours)
- Filter by radius
- Aggregate values across the time window for each track
- Compute Pearson correlation coefficients
- **NEW:** Include MCS-specific statistics (e.g., total_total_rain, duration_hours, total_area) in correlation analysis
- Optional masking of upper triangle (to avoid redundancy)
- Customizable appearance (colormap, annotations, size)

**Parameters:**
- `var_dfs_dict`: Dictionary of variable DataFrames (e.g., `var_dfs_scream`)
- `time_window`: Tuple of (start_hour, end_hour) relative to MCS initiation
- `radius`: Radius in degrees to analyze (default: 2.0)
- `mcs_stats_df`: (Optional) DataFrame with MCS statistics (e.g., `df_trop_scream_all`)
- `mcs_stats_cols`: (Optional) List of column names from mcs_stats_df to include (e.g., `['total_total_rain']`)
- `mask_upper`: Whether to mask upper triangle (default: True)
- `figsize`, `cmap`, `annot`, `fontsize`: Visualization options

**Returns:**
- `fig`: Matplotlib figure object
- `corr_matrix`: pandas DataFrame with correlation coefficients

```python
def remove_long_and_land(df_model, feature='duration_hours', p_high=0.9, 
                         lf_thresh=0.1,lf_var='total_land_fraction_track'):
    if df_model is None:
        return None
    q_high = df_model[feature].quantile(p_high)
    df_filtered = df_model[(df_model[feature] <= q_high)].copy()
    final_df = df_filtered[df_filtered[lf_var] <= lf_thresh].copy()
    return final_df
```

```python
# --- CUSTOM PLOTTING FUNCTION (Adapted for Rectangular Data) ---
def custom_rectangular_heatmap(corr_df, p_df, var_groups, group_colors, group_names, 
                      figsize=(12, 10), figname=None,
                      title='Correlogram', alpha_nonsig=0.4,
                      cmap='coolwarm', vmin=-0.4, vmax=0.4, p_thresh=0.05,
                      fontsize=12, title_fontsize=20, title_pos=0.92, 
                      cbar_label='Correlation Coefficient', border_columns=None):
    
    # Try to set font, fallback to default if not available
    try:
        mpl.rcParams['font.family'] = 'Helvetica'
    except:
        mpl.rcParams['font.family'] = 'sans-serif'
    
    # Set up the figure with white background
    fig = plt.figure(figsize=figsize, facecolor='white')
    
    # Create gridspec with left margin for group names
    # [left margin for group names, main plot, gap, colorbar]
    gs = gridspec.GridSpec(1, 4, figure=fig, width_ratios=[0.15, 1, 0.02, 0.05], wspace=0.05, hspace=0)
    
    # Main plot axis (skip the left margin)
    ax = fig.add_subplot(gs[0, 1])
    ax.set_facecolor('white')
    
    # Colorbar axis
    cax = fig.add_subplot(gs[0, 3])
    
    # Create colormap
    norm = mcolors.Normalize(vmin=vmin, vmax=vmax)
    try:
        cmap_obj = mpl.colormaps[cmap]
    except:
         cmap_obj = cm.get_cmap(cmap)
    
    # Get dimensions
    n_rows, n_cols = corr_df.shape
    
    # Create variable to group mapping for coloring labels
    var_to_color = {}
    for i, group in enumerate(var_groups):
        for var in group:
            var_to_color[var] = group_colors[i]
    
    # Plot the heatmap manually (Iterate Rows x Cols)
    for i in range(n_rows):
        for j in range(n_cols):
            
            # Get correlation and p-value
            corr_val = corr_df.iloc[i, j]
            p_val = p_df.iloc[i, j]
            
            # Skip if correlation or p-value is NaN
            if pd.isna(corr_val) or pd.isna(p_val):
                continue
                
            # Determine color and alpha based on significance
            color = cmap_obj(norm(corr_val))
            alpha = 1.0 if p_val < p_thresh else alpha_nonsig
            # For this summary, we force text black/white based on intensity or just black for readability
            text_color = 'gray' if abs(p_val) > 0 else 'black'
            
            # Plot the cell
            # y coordinate: n_rows - 1 - i to plot from top
            rect = patches.Rectangle((j, n_rows - 1 - i), 1, 1, facecolor=color, alpha=alpha, edgecolor='white', linewidth=1)
            ax.add_patch(rect)
            
            # Add correlation text
            ax.text(j + 0.5, n_rows - 1 - i + 0.5, f'{corr_val:.2f}', ha='center', va='center', color=text_color, 
                   fontsize=fontsize)
    
    # Set up the basic plot properties
    ax.set_xlim(0, n_cols)
    ax.set_ylim(0, n_rows)
    ax.set_aspect('equal') # Optional: remove if you want it to stretch to fill
    
    # Set ticks and labels
    col_names = corr_df.columns.tolist()
    row_names = corr_df.index.tolist()
    
    # X-axis labels (bottom) - Models
    ax.set_xticks(np.arange(n_cols) + 0.5)
    ax.set_xticklabels(col_names, rotation=45, ha='right', fontsize=fontsize, )
    
    # Y-axis labels (left side) - Variables
    # Need to reverse row names because we plot from bottom-up logic (0 is bottom in standard plot, but we mapped i=0 to top)
    # Actually, with (n_rows - 1 - i), i=0 is at y=n_rows-1 (top).
    # So we want the first label at the top.
    # set_yticks accepts positions. Top tick is at n_rows - 0.5
    ax.set_yticks(np.arange(n_rows) + 0.5)
    ax.set_yticklabels(row_names[::-1], fontsize=fontsize) # Reverse to match visual order
    
    # Color y-axis labels
    for tick_label, var_name in zip(ax.get_yticklabels(), row_names[::-1]):
        if var_name in var_to_color:
            tick_label.set_color(var_to_color[var_name])
           # tick_label.set_fontweight('bold')
    
    # Add horizontal lines to separate groups
    current_pos = 0
    for group in var_groups:
        current_pos += len(group)
        if current_pos < n_rows:
            y_pos = n_rows - current_pos
            ax.axhline(y=y_pos, color='black', linewidth=4, alpha=1)

    # Add group labels in the left margin
    current_pos = 0
    for i, (group, group_name, color) in enumerate(zip(var_groups, group_names, group_colors)):
        group_size = len(group)
        # Calculate middle of group in data coords
        group_start = n_rows - current_pos
        group_end = n_rows - (current_pos + group_size)
        group_middle = (group_start + group_end) / 2
        
        # Convert to axis coordinates for the text placement
        y_axis_coord = group_middle / n_rows
        
        ax.text(-0.5, y_axis_coord, group_name, 
                ha='center', va='center', fontsize=fontsize*1.3, 
                 color=color, rotation=90,
                transform=ax.transAxes)
        
        current_pos += group_size
    
    # Set title
    fig.suptitle(title, fontsize=title_fontsize, y=title_pos)
    
    # Remove spines
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)
    
    # Colorbar
    cbar = plt.colorbar(cm.ScalarMappable(norm=norm, cmap=cmap_obj), cax=cax)
    cbar.set_label(cbar_label, fontsize=12)
    cbar.ax.tick_params(labelsize=11)
    
    # Add border around specified columns (ERA5)
    if border_columns is not None and border_columns > 0:
        # Rectangle(xy, width, height)
        # We want to cover column 0 to border_columns, and all rows (0 to n_rows)
        border_rect = patches.Rectangle((0.0, 0.0), border_columns, n_rows, linewidth=4, edgecolor='black', 
                                      facecolor='none', zorder=100, clip_on=False)
        ax.add_patch(border_rect)
    
    # Save
    if figname:
        fig.savefig(figname, bbox_inches='tight', dpi=300, facecolor='w')
        print(f"Saved: {figname}")
    
    return fig, ax
```
